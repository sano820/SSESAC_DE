## 📚 1. 클라우드 컴퓨팅이란?

### 1.1 정의

**클라우드 컴퓨팅**: 인터넷을 통해 서버, 스토리지, 데이터베이스 등 IT 리소스를 필요할 때 필요한 만큼 사용하는 서비스

```
📊 전통적 방식 vs 클라우드

[전통적 방식 - 온프레미스(On-Premise)]
┌─────────────────────────────────────┐
│  🏢 자체 데이터센터                  │
│  ├── 서버 구매 💰💰💰              │
│  ├── 설치 및 설정 (2-3개월)         │
│  ├── 유지보수 인력 필요             │
│  └── 확장 시 추가 구매 필요          │
└─────────────────────────────────────┘

[클라우드 방식]
┌─────────────────────────────────────┐
│  ☁️ AWS/GCP/Azure 데이터센터       │
│  ├── 사용한 만큼만 지불 💰          │
│  ├── 몇 분 내 시작 가능             │
│  ├── 관리형 서비스 제공             │
│  └── 버튼 클릭으로 확장             │
└─────────────────────────────────────┘
```

### 1.2 클라우드 서비스 모델

![IaaS vs PaaS vs Saas 책임 범위 비교](https://www.redhat.com/rhdc/managed-files/styles/default_800/private/iaas-paas-saas-diagram5.1-1638x1046.png.webp?itok=1I1jAjm0)

| 모델 | 설명 | AWS 예시 | 사용자 책임 |
|------|------|----------|-------------|
| **IaaS** (Infrastructure as a Service) | 가상 서버, 네트워크 제공 | EC2, S3, VPC | OS, 애플리케이션 관리 |
| **PaaS** (Platform as a Service) | 관리형 플랫폼 제공 | MSK (Kafka), EMR (Spark), OpenSearch (Elasticsearch), MWAA (Airflow) | 애플리케이션 코드만 |
| **SaaS** (Software as a Service) | 완성된 소프트웨어 제공 | QuickSight (시각화), Athena (SQL 쿼리) | 사용만 |


### 1.3 클라우드의 장점

| 장점 | 설명 |
|------|------|
| **초기 비용 절감** | 서버 구매 불필요, 사용량 기반 과금 |
| **탄력성(Elasticity)** | 트래픽에 따라 자동 확장/축소 |
| **글로벌 배포** | 전 세계 데이터센터에 몇 분 내 배포 |
| **고가용성** | 다중 가용 영역으로 장애 대응 |
| **보안** | AWS가 물리적 보안, 규정 준수 담당 |
| **관리형 서비스** | 운영 부담 감소, 개발에 집중 |

## 📚 2. AWS 소개

### 2.1 AWS(Amazon Web Services)란?

- 2006년 시작된 세계 최대 클라우드 서비스 제공업체
- 200개 이상의 서비스 제공
- 전 세계 30개 이상의 리전(Region) 운영
- 시장 점유율 1위 (약 32%, 2024년 기준)

📊 주요 클라우드 제공업체

| **순위** | **플랫폼** | **사용률/점유율 (설문 기준)** | **주요 특징 및 트렌드** |
| --- | --- | --- | --- |
| **1위** | **AWS** (Amazon) | **약 48~52%** | 가장 넓은 서비스 생태계, 스타트업부터 대기업까지 표준화된 선택지 |
| **2위** | **Azure** (Microsoft) | **약 26~30%** | **성장세 가속.** OpenAI 서비스 통합 및 MS 생태계(Teams 등)와의 연동 강점 |
| **3위** | **GCP** (Google) | **약 22~25%** | 데이터 분석, AI/ML, Kubernetes(GKE) 기반의 기술 지향적 개발자 선호 |
| 4위 | **DigitalOcean** | 약 7~10% | 인디 개발자 및 소규모 스타트업의 높은 선호도 (사용 편의성) |
| 5위 | **Firebase** | 약 7~9% | 모바일/웹 프론트엔드 개발자들의 백엔드 서비스로 인기 |

### 2.2 AWS 글로벌 인프라

AWS는 전 세계에 분산된 인프라를 통해 서비스를 제공합니다. 이 구조를 이해하면 **어디에 서비스를 배포할지**, **어떻게 \*고가용성을 확보할지** 결정할 수 있습니다.
*고가용성: 시스템이 중단 없이(또는 아주 짧은 중단만 허용하고) 계속해서 서비스를 정상적으로 제공할 수 있는 상태*


```
🌍 AWS 글로벌 인프라 계층 구조

![AWS Global Infra](https://blog.kakaocdn.net/dna/dDDjBY/btsudQgH0uq/AAAAAAAAAAAAAAAAAAAAAEie8pk-yhtlpyRAl9wHoNzjezEgcIPjfpL4rU0fHrJL/img.png?credential=yqXZFxpELC7KVnFOS48ylbz2pIh7yKj8&expires=1769871599&allow_ip=&allow_referer=&signature=8YVnJFoq60RMmUp43CuzAaI6UTA%3D)

AWS 글로벌 인프라
│
├── Region (리전) ─────────────────── 가장 큰 단위
│   │
│   ├── Availability Zone (AZ) ───── 리전 내 독립 데이터센터
│   │   │
│   │   └── Data Center ──────────── 실제 물리 서버가 있는 건물
│   │
│   └── (여러 AZ가 하나의 리전 구성)
│
└── Edge Location ─────────────────── 전 세계 캐싱 포인트 (CDN)
```

- CDN: 물리적 거리로 인한 데이터 전송 지연(Latency)을 해결하기 위해 설계된 분산 네트워크 시스템

![CDN이란 무엇인가?](https://velog.velcdn.com/images%2Fyoungblue%2Fpost%2Ff29d7267-0498-4319-b72d-f17b7944b243%2F99CE4C415BF2B5FB18.png)

CDN의 데이터 처리 프로세스
사용자가 웹사이트에 접속하여 이미지나 동영상을 요청하면 다음과 같은 단계로 진행됩니다.
**사용자 요청:** 사용자의 요청이 오리진 서버로 바로 가지 않고, **가장 가까운 엣지 로케이션**으로 연결됩니다.
**캐시 확인 (Cache Hit/Miss):**
- **Hit:** 해당 엣지에 데이터가 있으면 즉시 사용자에게 전송합니다. (매우 빠름)
- **Miss:** 데이터가 없으면 엣지가 오리진 서버에 가서 데이터를 가져온 뒤, 사용자에게 전달하고 동시에 **복사본을 저장**합니다.
**이후 요청:** 이후 주변의 다른 사용자가 같은 데이터를 요청하면, 이미 저장된 복사본(캐시)을 즉시 내보냅니다.


#### 2.2.1 Region (리전)

**리전**은 지리적으로 분리된 AWS 데이터센터들의 클러스터입니다.

```
🗺️ 주요 리전 예시

북미                 유럽                  아시아 태평양
├── us-east-1       ├── eu-west-1        ├── ap-northeast-2 (서울) ⭐
│   (버지니아)      │   (아일랜드)       │
├── us-west-2       ├── eu-central-1     ├── ap-northeast-1 (도쿄)
│   (오레곤)        │   (프랑크푸르트)   │
└── ...             └── ...              └── ap-southeast-1 (싱가포르)
```

**리전 선택 기준:**

| 기준 | 설명 | 예시 |
|------|------|------|
| **지연 시간** | 사용자와 가까울수록 빠름 | 한국 서비스 → 서울 리전 |
| **규정 준수** | 데이터가 특정 국가에 있어야 함 | 개인정보보호법 → 국내 리전 |
| **서비스 가용성** | 모든 서비스가 모든 리전에 없음 | 신규 서비스 → 버지니아 먼저 |
| **비용** | 리전마다 가격이 다름 | 버지니아가 보통 가장 저렴 |

#### 2.2.2 Availability Zone (가용 영역, AZ)

**가용 영역**은 리전 내에서 물리적으로 분리된 독립 데이터센터입니다.

```
🏢 서울 리전(ap-northeast-2)의 가용 영역 구조

┌─────────────────── ap-northeast-2 (서울 리전) ───────────────────┐
│                                                                │
│   ┌─────────┐    ┌─────────┐    ┌─────────┐    ┌─────────┐     │
│   │   2a    │    │   2b    │    │   2c    │    │   2d    │     │
│   │ ┌─────┐ │    │ ┌─────┐ │    │ ┌─────┐ │    │ ┌─────┐ │     │
│   │ │ DC  │ │    │ │ DC  │ │    │ │ DC  │ │    │ │ DC  │ │     │
│   │ └─────┘ │    │ └─────┘ │    │ └─────┘ │    │ └─────┘ │     │
│   └────┬────┘    └────┬────┘    └────┬────┘    └────┬────┘     │
│        │              │              │              │          │
│        └──────────────┴──────────────┴──────────────┘          │
│                    고속 전용 네트워크 연결                          │
│              (낮은 지연시간, 높은 대역폭)                            │
└────────────────────────────────────────────────────────────────┘

각 AZ는:
- 독립된 전력, 냉각, 물리적 보안
- 서로 수십 km 떨어져 있음 (자연재해 대비)
- 전용 광케이블로 연결 (1ms 미만 지연)
```

**왜 다중 AZ를 사용할까?**

```
⚡ 단일 AZ vs 다중 AZ 비교

[단일 AZ 배포] - 위험!
┌─────────────┐
│     2a      │
│  ┌───────┐  │       AZ 장애 발생 시
│  │  EC2  │  │  ──────────────────>  서비스 전체 중단!
│  │  RDS  │  │
│  └───────┘  │
└─────────────┘

[다중 AZ 배포] - 권장!
┌─────────────┐     ┌─────────────┐
│     2a      │     │     2c      │       2a 장애 발생 시
│  ┌───────┐  │     │  ┌───────┐  │  ───────────────────>  2c가 처리!
│  │ EC2-1 │  │ ←─→ │  │ EC2-2 │  │       서비스 계속 유지
│  │ RDS   │  │     │  │ RDS   │  │
│  │(주)    │  │     │  │(대기)  │  │
│  └───────┘  │     │  └───────┘  │
└─────────────┘     └─────────────┘
```

#### 2.2.3 Edge Location (엣지 로케이션)

**엣지 로케이션**은 사용자에게 콘텐츠를 빠르게 전달하기 위한 캐싱 서버입니다.

```
🌐 Edge Location 동작 원리

[원본 서버가 서울에 있고, 미국 사용자가 접근하는 경우]

Edge Location 없이:
미국 사용자 ───────────────────────────────────> 서울 S3
             (태평양 횡단, 200ms+ 지연)

Edge Location 사용 (CloudFront):
                   ┌─────────────┐
                   │ 서울 S3      │ (원본)
                   │ (Origin)    │
                   └──────┬──────┘
                          │ 최초 1회만 전송
                          ▼
             ┌────────────────────────┐
             │  미국 Edge Location    │ (캐시 저장)
             └───────────┬────────────┘
                         │ 10ms 지연
                         ▼
                   미국 사용자
```

| 구성 요소 | 개수 (2024) | 주요 용도 |
|-----------|-------------|-----------|
| Region | 33개 | 서비스 배포, 데이터 저장 |
| Availability Zone | 105개+ | 고가용성 구성 |
| Edge Location | 600개+ | CDN 캐싱, 빠른 콘텐츠 전달 |

```
📍 한국 사용자를 위한 리전 선택

[ap-northeast-2] 서울 리전 ⭐ 추천
├── 한국에서 가장 낮은 지연 시간 (수 ms)
├── 4개의 가용 영역 (2a, 2b, 2c, 2d)
├── 대부분의 AWS 서비스 지원
└── 한국 데이터 규정 준수 가능

[ap-northeast-1] 도쿄 리전
├── 서울에 없는 일부 서비스 사용 시
└── 지연 시간: 약 30-50ms
```

## 📚 4. IAM (Identity and Access Management)

### 4.1 IAM이란?

**IAM**: AWS 리소스에 대한 접근을 안전하게 제어하는 서비스

```
🔐 IAM 핵심 개념

[Root Account] ─── 계정 생성 시 만들어지는 최고 권한 계정
      │            절대 일상 작업에 사용 금지!
      │
      ├── [IAM User] ─── 개별 사용자 계정
      │         │        사람 또는 애플리케이션용
      │         │
      │         └── [Access Key] ─── 프로그래밍 방식 접근용
      │
      ├── [IAM Group] ─── 사용자 묶음
      │         │         그룹에 정책 부여 → 소속 사용자에 적용
      │         │

      │

      │         │       AWS 서비스나 외부 사용자에게 권한 위임
      │         │
      │         └── 예: EC2가 S3에 접근할 수 있는 Role
      │
      └── [IAM Policy] ─── 권한 정의 (JSON)
                │         누가 어떤 리소스에 무엇을 할 수 있는지
                │
                └── 예: S3 읽기 전용, EC2 전체 권한
```

### 4.2 IAM Policy 구조

```json
{
    "Version": "2012-10-17",
    "Statement": [
        {
            "Effect": "Allow",           // Allow 또는 Deny
            "Action": [                  // 허용할 작업
                "s3:GetObject",
                "s3:ListBucket"

            "Resource": [                // 대상 리소스
                "arn:aws:s3:::my-bucket",

    ]
}
```

### 4.3 버킷 목록 조회

````
def list_buckets():
    """모든 S3 버킷 목록 조회"""
    try:
        response = s3_client.list_buckets()

        print("=" * 50)
        print("📦 S3 버킷 목록")
        print("=" * 50)

        for bucket in response["Buckets"]:
            print(f"🪣 {bucket['Name']}")
            print(f"   생성일: {bucket['CreationDate']}")
            print()

        return response["Buckets"]

    except ClientError as e:
        print(f"오류 발생: {e}")
        return []


# 실행
# buckets = list_buckets()
````

### 4.4 버킷 생성

````
def create_bucket(bucket_name: str, region: str = "ap-northeast-2"):
    """
    S3 버킷 생성

    Args:
        bucket_name: 버킷 이름 (전 세계 고유해야 함)
        region: AWS 리전
    """
    try:
        # ap-northeast-2 등 us-east-1 외 리전은 LocationConstraint 필요
        if region != "us-east-1":
            s3_client.create_bucket(
                Bucket=bucket_name,
                CreateBucketConfiguration={"LocationConstraint": region},
            )
        else:
            s3_client.create_bucket(Bucket=bucket_name)

        print(f"✅ 버킷 '{bucket_name}' 생성 완료!")
        return True

    except ClientError as e:
        error_code = e.response["Error"]["Code"]
        if error_code == "BucketAlreadyOwnedByYou":
            print(f"ℹ️ 버킷 '{bucket_name}'은 이미 존재합니다.")
        elif error_code == "BucketAlreadyExists":
            print(f"❌ 버킷 이름 '{bucket_name}'은 이미 사용 중입니다.")
        else:
            print(f"❌ 오류: {e}")
        return False


# 실행 (고유한 버킷 이름 사용!)
# create_bucket('my-unique-bucket-name-12345')
````

### 4.5 객체 업로드

````
def upload_file(bucket_name: str, file_path: str, object_key: str = None):
    """
    파일을 S3에 업로드

    Args:
        bucket_name: 버킷 이름
        file_path: 로컬 파일 경로
        object_key: S3에서의 키 (경로), None이면 파일명 사용
    """
    if object_key is None:
        object_key = file_path.split("/")[-1]

    try:
        s3_client.upload_file(file_path, bucket_name, object_key)
        print(f"✅ 업로드 완료: s3://{bucket_name}/{object_key}")
        return True

    except FileNotFoundError:
        print(f"❌ 파일을 찾을 수 없음: {file_path}")
        return False
    except ClientError as e:
        print(f"❌ 업로드 실패: {e}")
        return False


def upload_string(bucket_name: str, content: str, object_key: str):
    """
    문자열 데이터를 S3에 직접 업로드

    Args:
        bucket_name: 버킷 이름
        content: 저장할 문자열
        object_key: S3에서의 키
    """
    try:
        s3_client.put_object(
            Bucket=bucket_name, Key=object_key, Body=content.encode("utf-8")
        )
        print(f"✅ 업로드 완료: s3://{bucket_name}/{object_key}")
        return True

    except ClientError as e:
        print(f"❌ 업로드 실패: {e}")
        return False


# 예시: JSON 데이터 업로드
# data = {"name": "John", "age": 30}
# upload_string('my-bucket', json.dumps(data), 'data/user.json')
````

### 4.6 객체 목록 조회

````
def list_objects(bucket_name: str, prefix: str = ""):
    """
    버킷 내 객체 목록 조회

    Args:
        bucket_name: 버킷 이름
        prefix: 필터링할 접두사 (폴더)
    """
    try:
        # paginator 사용 (1000개 이상일 때 자동 페이징)
        paginator = s3_client.get_paginator("list_objects_v2")

        print(f"📂 s3://{bucket_name}/{prefix}")
        print("-" * 50)

        total_size = 0
        object_count = 0

        for page in paginator.paginate(Bucket=bucket_name, Prefix=prefix):
            if "Contents" not in page:
                print("(비어 있음)")
                continue

            for obj in page["Contents"]:
                size_kb = obj["Size"] / 1024
                print(f"  📄 {obj['Key']}")
                print(f"     크기: {size_kb:.2f} KB")
                print(f"     수정일: {obj['LastModified']}")

                total_size += obj["Size"]
                object_count += 1

        print("-" * 50)
        print(f"총 {object_count}개 객체, {total_size / 1024:.2f} KB")

    except ClientError as e:
        print(f"❌ 오류: {e}")


# 실행
list_objects('my-bucket', 'data/')
````

### 4.7 객체 다운로드

````
def download_file(bucket_name: str, object_key: str, local_path: str = None):
    """
    S3 객체 다운로드

    Args:
        bucket_name: 버킷 이름
        object_key: S3 객체 키
        local_path: 저장할 로컬 경로
    """
    if local_path is None:
        local_path = object_key.split("/")[-1]

    try:
        s3_client.download_file(bucket_name, object_key, local_path)
        print(f"✅ 다운로드 완료: {local_path}")
        return True

    except ClientError as e:
        if e.response["Error"]["Code"] == "404":
            print(f"❌ 객체를 찾을 수 없음: {object_key}")
        else:
            print(f"❌ 다운로드 실패: {e}")
        return False


def read_object(bucket_name: str, object_key: str) -> str:
    """
    S3 객체 내용을 문자열로 읽기

    Args:
        bucket_name: 버킷 이름
        object_key: S3 객체 키

    Returns:
        객체 내용 (문자열)
    """
    try:
        response = s3_client.get_object(Bucket=bucket_name, Key=object_key)
        content = response["Body"].read().decode("utf-8")
        return content

    except ClientError as e:
        print(f"❌ 읽기 실패: {e}")
        return None


# 실행
# content = read_object('my-bucket', 'data/user.json')
# if content:
#     data = json.loads(content)
#     print(data)
````

### 4.8 객체 삭제

````
def delete_object(bucket_name: str, object_key: str):
    """S3 객체 삭제"""
    try:
        s3_client.delete_object(Bucket=bucket_name, Key=object_key)
        print(f"✅ 삭제 완료: s3://{bucket_name}/{object_key}")
        return True
    except ClientError as e:
        print(f"❌ 삭제 실패: {e}")
        return False


def delete_prefix(bucket_name: str, prefix: str):
    """
    특정 prefix 아래 모든 객체 삭제

    ⚠️ 주의: 복구 불가능!
    """
    try:
        bucket = s3_resource.Bucket(bucket_name)
        objects = bucket.objects.filter(Prefix=prefix)

        delete_list = [{"Key": obj.key} for obj in objects]

        if not delete_list:
            print("삭제할 객체가 없습니다.")
            return

        # 최대 1000개씩 삭제
        response = s3_client.delete_objects(
            Bucket=bucket_name, Delete={"Objects": delete_list}
        )

        deleted_count = len(response.get("Deleted", []))
        print(f"✅ {deleted_count}개 객체 삭제 완료")

    except ClientError as e:
        print(f"❌ 삭제 실패: {e}")


# 실행
# delete_object('my-bucket', 'data/user.json')
# delete_prefix('my-bucket', 'temp/')
````

### 5.1 Presigned URL (미리 서명된 URL)

인증 없이 제한된 시간 동안 S3 객체에 접근할 수 있는 URL 생성

````
def generate_presigned_url(bucket_name: str, object_key: str, expiration: int = 3600):
    """
    Presigned URL 생성

    Args:
        bucket_name: 버킷 이름
        object_key: 객체 키
        expiration: 유효 시간 (초), 기본 1시간

    Returns:
        Presigned URL 문자열
    """
    try:
        url = s3_client.generate_presigned_url(
            "get_object",
            Params={"Bucket": bucket_name, "Key": object_key},
            ExpiresIn=expiration,
        )
        print(f"✅ Presigned URL 생성 (유효: {expiration}초)")
        print(f"   {url[:80]}...")
        return url

    except ClientError as e:
        print(f"❌ 생성 실패: {e}")
        return None


# 사용 예시
# url = generate_presigned_url('my-bucket', 'data/report.pdf', 3600)
````

````
# 필요한 라이브러리 설치
!pip install pandas s3fs

import pandas as pd


def save_df_to_s3(
    df: pd.DataFrame, bucket_name: str, key: str, file_format: str = "csv"
):
    """
    DataFrame을 S3에 저장

    Args:
        df: 저장할 DataFrame
        bucket_name: 버킷 이름
        key: S3 키
        file_format: 'csv' 또는 'parquet'
    """
    s3_path = f"s3://{bucket_name}/{key}"

    try:
        if file_format == "csv":
            df.to_csv(s3_path, index=False)
        elif file_format == "parquet":
            df.to_parquet(s3_path, index=False)
        else:
            raise ValueError(f"지원하지 않는 형식: {file_format}")

        print(f"✅ 저장 완료: {s3_path}")

    except Exception as e:
        print(f"❌ 저장 실패: {e}")


def load_df_from_s3(
    bucket_name: str, key: str, file_format: str = "csv"
) -> pd.DataFrame:
    """
    S3에서 DataFrame 로드

    Args:
        bucket_name: 버킷 이름
        key: S3 키
        file_format: 'csv' 또는 'parquet'

    Returns:
        DataFrame
    """
    s3_path = f"s3://{bucket_name}/{key}"

    try:
        if file_format == "csv":
            df = pd.read_csv(s3_path)
        elif file_format == "parquet":
            df = pd.read_parquet(s3_path)
        else:
            raise ValueError(f"지원하지 않는 형식: {file_format}")

        print(f"✅ 로드 완료: {s3_path}")
        print(f"   Shape: {df.shape}")
        return df

    except Exception as e:
        print(f"❌ 로드 실패: {e}")
        return None


# 예시
# df = pd.DataFrame({'name': ['Alice', 'Bob'], 'age': [25, 30]})
# save_df_to_s3(df, 'my-bucket', 'data/users.csv')
# df_loaded = load_df_from_s3('my-bucket', 'data/users.csv')
````

## 📚 6. boto3로 EC2 관리

````
import boto3
from botocore.exceptions import ClientError

# EC2 클라이언트 및 리소스 생성
ec2_client = boto3.client("ec2", region_name="ap-northeast-2")
ec2_resource = boto3.resource("ec2", region_name="ap-northeast-2")

print("EC2 클라이언트 생성 완료!")
````

### 6.1 인스턴스 목록 조회

````
def list_instances():
    """모든 EC2 인스턴스 목록 조회"""
    try:
        response = ec2_client.describe_instances()

        print("=" * 60)
        print("🖥️ EC2 인스턴스 목록")
        print("=" * 60)

        for reservation in response["Reservations"]:
            for instance in reservation["Instances"]:
                instance_id = instance["InstanceId"]
                instance_type = instance["InstanceType"]
                state = instance["State"]["Name"]

                # 이름 태그 찾기
                name = "N/A"
                for tag in instance.get("Tags", []):
                    if tag["Key"] == "Name":
                        name = tag["Value"]
                        break

                # Public IP (없을 수 있음)
                public_ip = instance.get("PublicIpAddress", "N/A")

                print(f"인스턴스 ID: {instance_id}")
                print(f"  이름: {name}")
                print(f"  유형: {instance_type}")
                print(f"  상태: {state}")
                print(f"  Public IP: {public_ip}")
                print("-" * 40)

    except ClientError as e:
        print(f"❌ 오류: {e}")


# 실행
# list_instances()
````

### 6.2 인스턴스 생성

````
def create_instance(
    name: str,
    ami_id: str = "ami-0c9c942bd7bf113a2",  # Amazon Linux 2023 (서울)
    instance_type: str = "t2.micro",
    key_name: str = None,
    security_group_ids: list = None,
):
    """
    EC2 인스턴스 생성

    Args:
        name: 인스턴스 이름
        ami_id: AMI ID
        instance_type: 인스턴스 유형
        key_name: 키 페어 이름
        security_group_ids: 보안 그룹 ID 리스트

    Returns:
        생성된 인스턴스 ID
    """
    try:
        params = {
            "ImageId": ami_id,
            "InstanceType": instance_type,
            "MinCount": 1,
            "MaxCount": 1,
            "TagSpecifications": [
                {"ResourceType": "instance", "Tags": [{"Key": "Name", "Value": name}]}
            ],
        }

        if key_name:
            params["KeyName"] = key_name

        if security_group_ids:
            params["SecurityGroupIds"] = security_group_ids

        response = ec2_client.run_instances(**params)

        instance_id = response["Instances"][0]["InstanceId"]
        print(f"✅ 인스턴스 생성 중: {instance_id}")
        print("   상태가 'running'이 될 때까지 잠시 대기하세요.")

        return instance_id

    except ClientError as e:
        print(f"❌ 생성 실패: {e}")
        return None


# 실행 (주의: 비용 발생 가능)
# instance_id = create_instance('my-test-instance', key_name='my-key')
````

### 6.3 인스턴스 시작/중지/종료

````
def start_instance(instance_id: str):
    """인스턴스 시작"""
    try:
        ec2_client.start_instances(InstanceIds=[instance_id])
        print(f"✅ 인스턴스 시작: {instance_id}")
    except ClientError as e:
        print(f"❌ 시작 실패: {e}")


def stop_instance(instance_id: str):
    """인스턴스 중지 (EBS 볼륨 유지)"""
    try:
        ec2_client.stop_instances(InstanceIds=[instance_id])
        print(f"✅ 인스턴스 중지: {instance_id}")
    except ClientError as e:
        print(f"❌ 중지 실패: {e}")


def terminate_instance(instance_id: str):
    """
    인스턴스 종료 (삭제)

    ⚠️ 주의: 복구 불가능! EBS 볼륨도 삭제됨 (기본 설정)
    """
    try:
        ec2_client.terminate_instances(InstanceIds=[instance_id])
        print(f"✅ 인스턴스 종료: {instance_id}")
    except ClientError as e:
        print(f"❌ 종료 실패: {e}")


# 실행
# start_instance('i-0123456789abcdef0')
# stop_instance('i-0123456789abcdef0')
# terminate_instance('i-0123456789abcdef0')
````

### 6.4 인스턴스 상태 대기

````
def wait_for_instance_running(instance_id: str):
    """인스턴스가 running 상태가 될 때까지 대기"""
    print(f"인스턴스 {instance_id} 시작 대기 중...")

    waiter = ec2_client.get_waiter("instance_running")
    waiter.wait(InstanceIds=[instance_id])

    # Public IP 조회
    response = ec2_client.describe_instances(InstanceIds=[instance_id])
    instance = response["Reservations"][0]["Instances"][0]
    public_ip = instance.get("PublicIpAddress", "N/A")

    print(f"✅ 인스턴스 시작 완료!")
    print(f"   Public IP: {public_ip}")

    return public_ip


# 실행
# public_ip = wait_for_instance_running('i-0123456789abcdef0')
````

## 📚 1. 네트워크 기초 (VPC 이해를 위한 사전 지식)

> VPC를 이해하려면 기본적인 네트워크 개념이 필요합니다. 핵심만 빠르게 살펴봅시다!

### 1.1 IP 주소란?

**IP 주소**: 네트워크에 연결된 모든 장치를 식별하는 고유 주소

![IP 주소 이진수 변환](https://ko.ipshu.com/ipv4-structure/202.103.0.68.png)

```
📊 IP 주소 구조 (IPv4)

192.168.1.100
 │   │   │  │
 └───┴───┴──┴── 4개의 숫자를 점(.)으로 구분

각 숫자는 0~255 범위 (8비트 = 1바이트)
총 32비트 = 약 43억 개의 고유 주소
```

IP 주소는 **네트워크 ID**와 **호스트 ID**로 구성됩니다.

```
📊 네트워크 ID vs 호스트 ID

IP 주소: 192.168.10.100
         ├────────┤ ├──┤
         네트워크 ID  호스트 ID
         (어느 네트워크?) (어느 장치?)

비유: 아파트 주소
├── 네트워크 ID = "강남구 테헤란로 123"  (건물 위치)
└── 호스트 ID = "1502호"                (세대 번호)
```

### 1.2 사설 IP vs 공용 IP

| 구분 | 사설 IP (Private) | 공용 IP (Public) |
|------|-------------------|------------------|
| **용도** | 내부 네트워크 전용 | 인터넷 통신용 |
| **비유** | 회사 내선 번호 | 회사 대표 전화번호 |
| **특징** | 외부에서 접근 불가 | 전 세계에서 유일 |
| **예시** | 192.168.x.x, 10.x.x.x | 52.78.xxx.xxx |

```
📊 AWS VPC에서 사용 가능한 사설 IP 대역

10.0.0.0/8      → 10.0.0.0 ~ 10.255.255.255    (대규모)
172.16.0.0/12   → 172.16.0.0 ~ 172.31.255.255  (AWS 기본 VPC)
192.168.0.0/16  → 192.168.0.0 ~ 192.168.255.255 (소규모)
```

> 💡 VPC는 **사설 IP 대역**을 사용합니다. 인터넷과 통신하려면 공용 IP가 필요합니다.

### 1.3 CIDR 표기법

**CIDR** (Classless Inter-Domain Routing): IP 주소 범위를 간결하게 표현하는 방법

![CIDR 표기법](https://blog.kakaocdn.net/dn/bkCqnT/btsGQTgRnJe/c8OrdmMRLUKTgKo1DYf4Xk/img.png)

```
📊 CIDR 표기법 이해하기

192.168.1.0/24
             ├─ /24 = 앞에서 24비트가 네트워크 ID
             └─ 나머지 8비트(32-24)가 호스트 ID

/24 → 2^8 = 256개 IP 사용 가능
/16 → 2^16 = 65,536개 IP 사용 가능
/8  → 2^24 = 16,777,216개 IP 사용 가능
```

| CIDR | 서브넷 마스크 | 사용 가능 IP | 용도 |
|------|--------------|-------------|------|
| /32 | 255.255.255.255 | 1개 | 단일 호스트 |
| /24 | 255.255.255.0 | 256개 | 소규모 서브넷 |
| /16 | 255.255.0.0 | 65,536개 | VPC 기본 크기 |
| /8 | 255.0.0.0 | 16,777,216개 | 대규모 네트워크 |

> 💡 **숫자가 작을수록 더 많은 IP** → `/16`이 `/24`보다 큼

### 1.4 서브넷 (Subnet)

**서브넷**: 큰 네트워크를 작은 네트워크로 나눈 것

```
📊 서브넷 분할 예시

VPC: 10.0.0.0/16 (65,536개 IP)
│
├── 서브넷 A: 10.0.1.0/24 (256개 IP) → 웹 서버용
├── 서브넷 B: 10.0.2.0/24 (256개 IP) → 앱 서버용
└── 서브넷 C: 10.0.3.0/24 (256개 IP) → DB 서버용

비유: 큰 사무실을 칸막이로 나눠 부서별로 배치
```

**AWS 서브넷 주의사항**: 각 서브넷에서 **5개 IP는 AWS가 예약**

```
예: 10.0.1.0/24 서브넷 (256개 IP)

10.0.1.0   → 네트워크 주소 (예약)
10.0.1.1   → VPC 라우터 (예약)
10.0.1.2   → AWS DNS (예약)
10.0.1.3   → 미래 사용 (예약)
10.0.1.255 → 브로드캐스트 (예약)

✅ 실제 사용 가능: 251개 (256 - 5)
```

## 📚 2. VPC (Virtual Private Cloud)

### 2.1 VPC란?

![VPC란](https://docs.aws.amazon.com/ko_kr/vpc/latest/userguide/images/how-it-works.png)

**VPC**: AWS 클라우드 내 논리적으로 격리된 가상 네트워크

- 온프레미스 데이터센터와 유사한 네트워크 환경
- IP 주소 범위, 서브넷, 라우팅, 보안 그룹 제어
- 다른 VPC와 완전히 격리됨

```
📊 VPC 기본 구조

┌─────────────────── VPC (10.0.0.0/16) ───────────────────┐
│                                                          │
│  ┌──────── 가용 영역 A ─────┐  ┌──────── 가용 영역 B ───┐     │
│  │                          │  │                        ││
│  │ ┌────────────────────┐   │  │ ┌────────────────────┐ ││
│  │ │ 퍼블릭 서브넷         │   │  │ │ 퍼블릭 서브넷          │ ││
│  │ │ 10.0.1.0/24        │   │  │ │ 10.0.2.0/24        │ ││
│  │ │  [EC2] [NAT GW]    │   │  │ │  [EC2]              │ ││
│  │ └────────────────────┘   │  │ └────────────────────┘ ││
│  │                          │  │                        ││
│  │ ┌────────────────────┐   │  │ ┌────────────────────┐ ││
│  │ │ 프라이빗 서브넷        │   │  │ │ 프라이빗 서브넷        │ ││
│  │ │ 10.0.3.0/24        │   │  │ │ 10.0.4.0/24        │ ││
│  │ │  [RDS] [EC2]       │   │  │ │  [RDS Standby]     │ ││
│  │ └────────────────────┘   │  │ └────────────────────┘ ││
│  │                          │  │                        ││
│  └──────────────────────────┘  └────────────────────────┘│
│                                                          │
│         ┌──────────────────┐                             │
│         │ Internet Gateway │ ←── 인터넷 연결                │
│         └──────────────────┘                             │
│                                                          │
└──────────────────────────────────────────────────────────┘
```

> 💡 **서브넷(Subnet)**: VPC 내 IP 주소 범위를 나눈 "작은 네트워크". 집을 방으로 나누듯, 용도별로 네트워크를 분리합니다.

![서브넷 이미지](https://velog.velcdn.com/images/sangmin1998/post/023978e9-a973-42ad-a840-d921dff24b4a/image.png)

### 2.2 VPC 핵심 구성 요소

| 구성 요소 | 설명 | 예시 |
|-----------|------|------|
| **VPC** | 가상 네트워크 (IP 주소 범위: CIDR) | `10.0.0.0/16` (65,536개 IP) |
| **서브넷** | VPC 내 IP 주소 범위를 나눈 것 | `10.0.1.0/24` (256개 IP) |
| **인터넷 게이트웨이 (IGW)** | VPC와 인터넷 연결 | `igw-0abc123` |
| **NAT 게이트웨이** | 프라이빗 서브넷 → 인터넷 (단방향) | 프라이빗 EC2가 패키지 업데이트 |
| **라우팅 테이블** | 트래픽 경로 정의 | `0.0.0.0/0 → igw` |
| **보안 그룹** | 인스턴스 레벨 방화벽 | SSH(22), HTTP(80) 허용 |
| **NACL** | 서브넷 레벨 방화벽 | 특정 IP 대역 차단 |


### 2.3 퍼블릭 vs 프라이빗 서브넷

```
📊 서브넷 유형 비교

[퍼블릭 서브넷]
├── 인터넷 *게이트웨이로 라우팅 O
├── Public IP 할당 가능
├── 외부에서 직접 접근 가능
└── 용도: 웹 서버, 배스천 호스트

[프라이빗 서브넷]
├── 인터넷 게이트웨이로 라우팅 X
├── 외부에서 직접 접근 불가
├── NAT Gateway 통해 아웃바운드만 가능
└── 용도: 데이터베이스, 내부 서버
```

*게이트웨이: 서로 다른 네트워크(VPC ↔ 인터넷) 사이를 연결해주는 출입구 역할을 하는 네트워크 장치

### 2.4 기본 VPC

- AWS 계정 생성 시 각 리전에 자동 생성
- 바로 사용 가능한 기본 네트워크 환경
- CIDR: 172.31.0.0/16
- 가용 영역별 퍼블릭 서브넷 자동 생성

> 💡 **실습에서는 기본 VPC 사용을 권장**합니다. 프로덕션에서는 사용자 정의 VPC 생성

## 📚 3. RDS (Relational Database Service)

### 3.1 RDS란?

**RDS**: AWS의 관리형 관계형 데이터베이스 서비스

- 데이터베이스 설치, 패치, 백업 자동화
- 고가용성 (Multi-AZ)
- 읽기 전용 복제본 (Read Replica)
- 자동 스케일링

```
📊 직접 관리 vs RDS

[EC2에 직접 DB 설치]
├── OS 패치 👈 사용자
├── DB 설치/업그레이드 👈 사용자
├── 백업 설정 👈 사용자
├── 복제 구성 👈 사용자
├── 모니터링 설정 👈 사용자
└── 장애 대응 👈 사용자

[RDS 사용]
├── OS 패치 ✅ AWS
├── DB 설치/업그레이드 ✅ AWS
├── 백업 설정 ✅ AWS (자동)
├── 복제 구성 ✅ AWS (클릭)
├── 모니터링 설정 ✅ AWS (CloudWatch)
└── 장애 대응 ✅ AWS (Multi-AZ 자동 장애조치)
```

### 3.2 지원 데이터베이스 엔진

| 엔진 | 설명 | Free Tier |
|------|------|-----------|
| **MySQL** | 가장 인기 있는 오픈소스 DB | ✅ db.t3.micro |
| **PostgreSQL** | 고급 기능의 오픈소스 DB | ✅ db.t3.micro |
| **MariaDB** | MySQL 호환 오픈소스 DB | ✅ db.t3.micro |
| **Oracle** | 엔터프라이즈 DB | ❌ 라이선스 비용 |
| **SQL Server** | Microsoft DB | ❌ 라이선스 비용 |
| **Aurora** | AWS 자체 개발, MySQL/PostgreSQL 호환 | ❌ |

### 3.3 RDS 인스턴스 클래스

```
📊 RDS 인스턴스 클래스 (Free Tier)

db.t3.micro / db.t4g.micro
├── vCPU: 2
├── 메모리: 1 GB
├── 네트워크: 최대 5 Gbps
└── 무료 사용량: 750시간/월 (12개월)

⚠️ 스토리지 20GB (gp2) 무료
⚠️ 백업 스토리지 20GB 무료
```

### 3.4 Multi-AZ 배포

```
📊 Multi-AZ 구성

[가용 영역 A]           [가용 영역 B]
┌─────────────┐         ┌─────────────┐
│  Primary    │ ──동기── │  Standby    │
│    RDS      │   복제   │    RDS      │
│  (활성)      │         │  (대기)      │

       │
       │ 장애 발생 시

[가용 영역 A]           [가용 영역 B]
┌─────────────┐         ┌─────────────┐
│  장애        │         │  Primary    │
│             │         │    RDS      │
│             │         │  (활성)      │
└─────────────┘         └─────────────┘
```

### 5.1 PostgreSQL 연결

````
import psycopg2
from sqlalchemy import create_engine
import pandas as pd

# RDS 연결 정보
RDS_HOST = "your-db-instance.xxxxxxxx.ap-northeast-2.rds.amazonaws.com"
RDS_PORT = 5432
RDS_DATABASE = "mydb"
RDS_USER = "admin"
RDS_PASSWORD = "your-password"  # 실제로는 환경 변수나 Secrets Manager 사용


def connect_postgresql():
    """PostgreSQL RDS 연결"""
    try:
        conn = psycopg2.connect(
            host=RDS_HOST,
            port=RDS_PORT,
            database=RDS_DATABASE,
            user=RDS_USER,
            password=RDS_PASSWORD,
        )
        print("✅ PostgreSQL 연결 성공!")
        return conn
    except Exception as e:
        print(f"❌ 연결 실패: {e}")
        return None


# conn = connect_postgresql()
````

### 5.2 SQLAlchemy 사용

````
def get_sqlalchemy_engine():
    """SQLAlchemy 엔진 생성"""
    connection_string = (
        f"postgresql://{RDS_USER}:{RDS_PASSWORD}@{RDS_HOST}:{RDS_PORT}/{RDS_DATABASE}"
    )

    engine = create_engine(connection_string)
    print("✅ SQLAlchemy 엔진 생성!")
    return engine


# engine = get_sqlalchemy_engine()

# DataFrame을 RDS에 저장
# df = pd.DataFrame({'name': ['Alice', 'Bob'], 'age': [25, 30]})
# df.to_sql('users', engine, if_exists='replace', index=False)

# RDS에서 DataFrame으로 읽기
# df_loaded = pd.read_sql('SELECT * FROM users', engine)
````

### 5.3 샘플 테이블 생성 및 데이터 삽입

````
def create_sample_table(conn):
    """샘플 테이블 생성"""
    cursor = conn.cursor()

    # 테이블 생성
    cursor.execute("""
        CREATE TABLE IF NOT EXISTS sales (
            id SERIAL PRIMARY KEY,
            product_name VARCHAR(100),
            quantity INTEGER,
            price DECIMAL(10,2),
            sale_date DATE DEFAULT CURRENT_DATE
        )
    """)

    # 샘플 데이터 삽입
    cursor.execute("""
        INSERT INTO sales (product_name, quantity, price)
        VALUES
            ('노트북', 10, 1500000),
            ('키보드', 50, 50000),
            ('마우스', 100, 30000),
            ('모니터', 20, 300000)
    """)

    conn.commit()
    print("✅ 테이블 생성 및 데이터 삽입 완료!")

    # 데이터 조회
    cursor.execute("SELECT * FROM sales")
    rows = cursor.fetchall()

    print("\n📊 sales 테이블:")
    for row in rows:
        print(f"  {row}")

    cursor.close()


# conn = connect_postgresql()
# create_sample_table(conn)
# conn.close()
````

### 6.1 RDS 인스턴스 목록 조회

````
def list_rds_instances():
    """모든 RDS 인스턴스 목록 조회"""
    try:
        response = rds_client.describe_db_instances()

        print("=" * 60)
        print("🗄️ RDS 인스턴스 목록")
        print("=" * 60)

        for db in response["DBInstances"]:
            print(f"인스턴스 ID: {db['DBInstanceIdentifier']}")
            print(f"  엔진: {db['Engine']} {db['EngineVersion']}")
            print(f"  클래스: {db['DBInstanceClass']}")
            print(f"  상태: {db['DBInstanceStatus']}")
            print(f"  엔드포인트: {db.get('Endpoint', {}).get('Address', 'N/A')}")
            print(f"  포트: {db.get('Endpoint', {}).get('Port', 'N/A')}")
            print("-" * 40)

    except Exception as e:
        print(f"❌ 오류: {e}")


# 실행
# list_rds_instances()
````

### 6.2 RDS 인스턴스 생성 (boto3)

````
def create_rds_instance(
    db_identifier: str, db_name: str, master_username: str, master_password: str
):
    """
    RDS PostgreSQL 인스턴스 생성

    Args:
        db_identifier: 인스턴스 식별자
        db_name: 초기 데이터베이스 이름
        master_username: 마스터 사용자 이름
        master_password: 마스터 비밀번호
    """
    try:
        response = rds_client.create_db_instance(
            DBInstanceIdentifier=db_identifier,
            DBName=db_name,
            Engine="postgres",
            EngineVersion="15.4",
            DBInstanceClass="db.t3.micro",
            MasterUsername=master_username,
            MasterUserPassword=master_password,
            AllocatedStorage=20,
            StorageType="gp2",
            PubliclyAccessible=True,  # 실습용
            BackupRetentionPeriod=0,  # 백업 비활성화 (Free Tier)
            Tags=[
                {"Key": "Name", "Value": db_identifier},
                {"Key": "Environment", "Value": "Development"},
            ],
        )

        print(f"✅ RDS 인스턴스 생성 중: {db_identifier}")
        print("   완료까지 5-10분 소요됩니다.")

        return response["DBInstance"]

    except Exception as e:
        print(f"❌ 생성 실패: {e}")
        return None


# 실행 (주의: 비용 발생 가능)
# create_rds_instance('my-postgres', 'mydb', 'admin', 'YourStrongPassword123!')
````

### 6.3 RDS 인스턴스 삭제

````
def delete_rds_instance(db_identifier: str, skip_final_snapshot: bool = True):
    """
    RDS 인스턴스 삭제

    ⚠️ 주의: 데이터가 모두 삭제됩니다!
    """
    try:
        rds_client.delete_db_instance(
            DBInstanceIdentifier=db_identifier,
            SkipFinalSnapshot=skip_final_snapshot,
            DeleteAutomatedBackups=True,
        )
        print(f"✅ RDS 인스턴스 삭제 중: {db_identifier}")

    except Exception as e:
        print(f"❌ 삭제 실패: {e}")


# 실행
# delete_rds_instance('my-postgres')
````

# Day 09 - 5교시: EC2에 Kafka 배포하고 로컬에서 연결하기

## 학습 목표

- EC2 인스턴스에 Docker Compose로 Kafka를 배포할 수 있다
- Security Group 설정으로 외부에서 Kafka에 접근할 수 있도록 구성한다
- KAFKA_ADVERTISED_LISTENERS 설정의 역할을 이해하고 외부 연결용 리스너를 구성한다
- 로컬(WSL2)에서 EC2의 Kafka에 연결하여 메시지를 주고받을 수 있다
- AWS 환경에서 서비스 운영의 기본 흐름을 체험한다

---

## 왜 이 실습이 중요한가요?

지금까지 우리는 로컬 환경(Docker Desktop)에서 Kafka를 사용했습니다.
하지만 실제 프로젝트에서는 AWS 같은 클라우드 환경에서 서비스를 운영합니다.

```

│  ┌───────────────┐  │            │  ┌───────────────┐  │
│  │    Docker     │  │            │  │    Docker     │  │
│  │    Kafka      │  │   ────▶    │  │    Kafka      │  │
│  │    Producer   │  │            │  │               │  │
│  │    Consumer   │  │            │  └───────────────┘  │
│  └───────────────┘  │            │                     │
└─────────────────────┘            └─────────────────────┘
                                             │
                                             │ 9093 포트
                                             ▼
                                   ┌─────────────────────┐
                                   │     내 PC (WSL2)     │
                                   │  Python Producer    │
                                   │  Python Consumer    │
```

이번 실습에서는:
1. EC2에 Kafka를 배포하고

3. 로컬에서 클라우드의 Kafka에 연결하는 경험을 합니다.

이 경험은 **2차 프로젝트**와 **최종 프로젝트**에서 AWS를 사용할 때 큰 도움이 됩니다!

EC2에서 Kafka 설정 파일을 만듭니다. **핵심은 ADVERTISED_LISTENERS 설정**입니다.

```bash
# 작업 디렉토리 생성
mkdir -p ~/kafka-ec2
cd ~/kafka-ec2
```

`compose.yml` 파일을 생성합니다:

```yaml
# ~/kafka-ec2/compose.yml
services:
  kafka:
    image: apache/kafka:latest
    container_name: kafka-broker
    ports:
      - "9092:9092"   # 내부용 (EC2 안에서만)
      - "9093:9093"   # 외부용 (로컬 PC에서 접근)
    environment:
      KAFKA_NODE_ID: 1
      KAFKA_PROCESS_ROLES: broker,controller
      KAFKA_CONTROLLER_QUORUM_VOTERS: 1@kafka:9094

      # ============================================
      # 리스너 설정 (핵심!)
      # ============================================
      # LISTENERS: Kafka가 실제로 바인딩하는 주소
      KAFKA_LISTENERS: INTERNAL://0.0.0.0:9092,EXTERNAL://0.0.0.0:9093,CONTROLLER://0.0.0.0:9094

      # ADVERTISED_LISTENERS: 클라이언트에게 알려줄 주소
      # - INTERNAL: Docker 내부 통신용 (컨테이너 이름)
      # - EXTERNAL: 외부(로컬 PC)에서 접근용 (EC2 Public IP)
      KAFKA_ADVERTISED_LISTENERS: INTERNAL://kafka:9092,EXTERNAL://${EC2_PUBLIC_IP}:9093

      KAFKA_LISTENER_SECURITY_PROTOCOL_MAP: INTERNAL:PLAINTEXT,EXTERNAL:PLAINTEXT,CONTROLLER:PLAINTEXT
      KAFKA_CONTROLLER_LISTENER_NAMES: CONTROLLER
      KAFKA_INTER_BROKER_LISTENER_NAME: INTERNAL

      # 토픽 자동 생성 및 기본 설정
      KAFKA_AUTO_CREATE_TOPICS_ENABLE: "true"
      KAFKA_NUM_PARTITIONS: 3
      KAFKA_DEFAULT_REPLICATION_FACTOR: 1
      KAFKA_OFFSETS_TOPIC_REPLICATION_FACTOR: 1
      KAFKA_TRANSACTION_STATE_LOG_REPLICATION_FACTOR: 1
      KAFKA_TRANSACTION_STATE_LOG_MIN_ISR: 1
    healthcheck:
      test: ["CMD-SHELL", "/opt/kafka/bin/kafka-broker-api-versions.sh --bootstrap-server localhost:9092 || exit 1"]
      interval: 10s
      timeout: 10s
      retries: 5
      start_period: 30s

  kafka-ui:
    image: provectuslabs/kafka-ui:latest
    container_name: kafka-ui
    ports:
      - "8080:8080"
    depends_on:
      kafka:
        condition: service_healthy
    environment:
      KAFKA_CLUSTERS_0_NAME: ec2-kafka
      KAFKA_CLUSTERS_0_BOOTSTRAPSERVERS: kafka:9092
```

### 3.5 KAFKA_ADVERTISED_LISTENERS 이해하기

이 설정이 왜 중요한지 그림으로 이해해봅시다.

```
❌ 문제 상황: ADVERTISED_LISTENERS가 kafka:9092만 있을 때

   [로컬 PC]                          [EC2]
   ┌──────────┐                      ┌──────────────────┐
   │ Producer │ ──── 1. 연결 ───────▶ │ Kafka (9093)     │
   │          │                      │                  │
   │          │◀── 2. 응답 ───────────│ "다음에는          │
   │          │    "kafka:9092로 와"  │  kafka:9092로 와" │
   │          │                      └──────────────────┘
   │          │
   │          │ ──── 3. 재연결 시도 ──▶ kafka:9092 ???
   │          │                         (알 수 없는 호스트!)
   └──────────┘

   결과: 첫 연결은 되지만, 메타데이터 갱신 시 연결 끊김!
```

```
✅ 해결책: 리스너 분리

   [로컬 PC]                          [EC2]
   ┌──────────┐                      ┌──────────────────┐
   │ Producer │ ──── 1. 연결 ────────▶│ Kafka (9093)     │
   │          │      (EXTERNAL)      │                  │
   │          │◀── 2. 응답 ───────────│ "다음에도          │
   │          │    "EC2_IP:9093으로"  │  EC2_IP:9093으로" │
   │          │                      └──────────────────┘
   │          │
   │          │ ──── 3. 재연결 ──────▶ EC2_IP:9093 ✅
   │          │                         (연결 유지!)
   └──────────┘

   ADVERTISED_LISTENERS 설정:
   - INTERNAL://kafka:9092     → 컨테이너 내부용
   - EXTERNAL://EC2_IP:9093    → 외부(로컬)용
```

**핵심 포인트**:
- Kafka 클라이언트는 첫 연결 후 브로커에게 "다음 연결 주소"를 받습니다
- 이 주소가 `ADVERTISED_LISTENERS`입니다
- 외부에서 접속할 때는 **EC2의 Public IP**를 알려줘야 합니다

### 4.2 Producer 코드 (로컬에서 실행)

로컬 PC에서 EC2의 Kafka로 메시지를 전송합니다.

````
# local_producer.py
# 이 코드를 로컬 PC(WSL2)에서 실행합니다

from confluent_kafka import Producer
import json

# =============================================================================
# EC2 Public IP 설정 (본인의 EC2 IP로 변경!)
# =============================================================================
EC2_PUBLIC_IP = "YOUR_EC2_PUBLIC_IP"  # 예: "13.125.xxx.xxx"
KAFKA_BOOTSTRAP_SERVERS = f"{EC2_PUBLIC_IP}:9093"


def delivery_callback(err, msg):
    """메시지 전송 결과 콜백"""
    if err:
        print(f"  전송 실패: {err}")
    else:
        print(f"  전송 성공: partition={msg.partition()}, offset={msg.offset()}")


# Producer 설정
config = {
    "bootstrap.servers": KAFKA_BOOTSTRAP_SERVERS,
    "client.id": "local-producer",
}

producer = Producer(config)

print(f"EC2 Kafka에 연결 중... ({KAFKA_BOOTSTRAP_SERVERS})")
print("-" * 50)

# 테스트 메시지 전송
for i in range(1, 6):
    message = {
        "message_id": i,
        "content": f"Hello from Local PC #{i}",
        "source": "WSL2 Ubuntu",
    }

    producer.produce(
        topic="test-from-local",
        key=f"msg-{i}".encode("utf-8"),
        value=json.dumps(message, ensure_ascii=False).encode("utf-8"),
        callback=delivery_callback,
    )
    print(f"메시지 {i} 버퍼에 추가")

# 모든 메시지 전송 완료 대기
producer.flush()
print("-" * 50)
print("모든 메시지 전송 완료!")
print(f"Kafka UI 확인: http://{EC2_PUBLIC_IP}:8080")
````

### 4.3 Consumer 코드 (로컬에서 실행)

로컬 PC에서 EC2의 Kafka로부터 메시지를 수신합니다.

````
# local_consumer.py
# 이 코드를 로컬 PC(WSL2)에서 실행합니다

from confluent_kafka import Consumer, KafkaError
import json
import signal

# =============================================================================
# EC2 Public IP 설정 (본인의 EC2 IP로 변경!)
# =============================================================================
EC2_PUBLIC_IP = "YOUR_EC2_PUBLIC_IP"  # 예: "13.125.xxx.xxx"
KAFKA_BOOTSTRAP_SERVERS = f"{EC2_PUBLIC_IP}:9093"

# Consumer 설정
config = {
    "bootstrap.servers": KAFKA_BOOTSTRAP_SERVERS,
    "group.id": "local-consumer-group",
    "client.id": "local-consumer",
    "auto.offset.reset": "earliest",  # 처음부터 읽기
    "enable.auto.commit": True,
}

consumer = Consumer(config)
running = True


def signal_handler(signum, frame):
    """Ctrl+C 처리"""
    global running
    print("\n종료 신호 수신...")
    running = False


signal.signal(signal.SIGINT, signal_handler)
signal.signal(signal.SIGTERM, signal_handler)

print(f"EC2 Kafka Consumer 시작 ({KAFKA_BOOTSTRAP_SERVERS})")
print("토픽: test-from-local")
print("Ctrl+C로 종료")
print("-" * 50)

consumer.subscribe(["test-from-local"])

try:
    while running:
        msg = consumer.poll(timeout=1.0)

        if msg is None:
            continue

        if msg.error():
            if msg.error().code() != KafkaError._PARTITION_EOF:
                print(f"에러: {msg.error()}")
            continue

        # 메시지 처리
        key = msg.key().decode("utf-8") if msg.key() else None
        value = json.loads(msg.value().decode("utf-8"))

        print(f"메시지 수신:")
        print(f"  Key: {key}")
        print(f"  Value: {value}")
        print(f"  Partition: {msg.partition()}, Offset: {msg.offset()}")
        print()

finally:
    consumer.close()
    print("Consumer 종료 완료")
````

- 8080: Kafka UI 웹 접근용

이 실습에서는 EXTERNAL 리스너를 9093으로 설정했으므로, 외부에서 Kafka에 연결하려면 9093 포트를 열어야 합니다.
</details>

### Q2. KAFKA_ADVERTISED_LISTENERS에서 EXTERNAL 리스너를 EC2 Public IP로 설정해야 하는 이유는?

<details>
<summary>정답 확인</summary>

Kafka 클라이언트가 첫 연결 후 브로커로부터 받는 "다음 연결 주소"가 ADVERTISED_LISTENERS이기 때문입니다.

만약 `kafka:9092`(내부 주소)만 설정하면:
1. 로컬 PC → EC2:9093 첫 연결 성공
2. 브로커가 "다음엔 kafka:9092로 와" 라고 알려줌
3. 로컬 PC는 `kafka`라는 호스트를 모르므로 연결 실패

EXTERNAL을 EC2 Public IP로 설정하면:
1. 로컬 PC → EC2:9093 연결
2. 브로커가 "다음엔 EC2_PUBLIC_IP:9093으로 와" 라고 알려줌
3. 로컬 PC가 계속 연결 유지 가능
</details>

A) EC2 인스턴스 중지(Stop)
B) Elastic IP가 있다면 해제
C) Security Group 삭제
D) 사용하지 않는 EBS 볼륨 삭제

- Security Group 자체는 무료이며, 삭제하면 다음 실습 때 다시 만들어야 함
</details>

### 과제 3: 실시간 센서 데이터 파이프라인 (난이도: ★★★)

**목표**: 간단한 센서 데이터 파이프라인 구현

**시나리오**:
- IoT 센서 3개가 1초마다 온도 데이터를 전송
- Consumer가 데이터를 받아서 이상 온도(30도 초과) 감지 시 경고 출력

**요구사항**:

3. Consumer: 30도 초과 시 "경고: sensor-xxx 온도 이상!" 출력
4. 10초간 실행 후 자동 종료

```python
# sensor_consumer.py
from confluent_kafka import Consumer, KafkaError
import json
import signal
import time

EC2_PUBLIC_IP = "YOUR_EC2_IP"

consumer = Consumer({
    'bootstrap.servers': f'{EC2_PUBLIC_IP}:9093',
    'group.id': 'sensor-monitor-group',
    'auto.offset.reset': 'latest',  # 최신 메시지부터
})

running = True
start_time = time.time()

def signal_handler(signum, frame):
    global running
    running = False

signal.signal(signal.SIGINT, signal_handler)

print("센서 모니터링 시작 (이상 온도 감지)")
print("=" * 50)

consumer.subscribe(['sensor-data'])
alert_count = 0

try:
    while running and (time.time() - start_time) < 15:  # 15초간 실행
        msg = consumer.poll(timeout=1.0)

        if msg is None:
            continue

        if msg.error():
            if msg.error().code() != KafkaError._PARTITION_EOF:
                print(f"에러: {msg.error()}")
            continue

        data = json.loads(msg.value().decode('utf-8'))
        sensor_id = data['sensor_id']
        temp = data['temperature']

        if temp > 30:
            alert_count += 1
            print(f"[경고 #{alert_count}] {sensor_id} 온도 이상! ({temp}C)")
        else:
            print(f"[정상] {sensor_id} = {temp}C")

finally:
    consumer.close()
    print("=" * 50)
    print(f"모니터링 완료: 총 {alert_count}건의 이상 온도 감지")
```

```python
from langchain_core.prompts import FewShotChatMessagePromptTemplate
```

```python
examples = [
    {"input": "지구의 대기 중 가장 많은 비율을 차지하는 기체는 무엇인가요?", "output": "질소입니다."},
    {"input": "광합성에 필요한 주요 요소들은 무엇인가요?", "output": "빛, 이산화탄소, 물입니다."},
]
```

```python
example_prompt = ChatPromptTemplate.from_messages(
    [
        ('human','{input}'),
        ('ai','{output}')
    ]
)
```

```python
few_shot_prompt = FewShotChatMessagePromptTemplate(
    example_prompt = example_prompt,
    examples = examples,
)
```

```python
final_prompt = ChatPromptTemplate.from_messages(
    [
        ('system','당신은 과학과 수학에 대해 잘 아는 학생이야. 아래 예시를 참고해서 답변을 줘.'),
        few_shot_prompt,
        ('human','{input}')
    ]
)
```

```python
llm = ChatOpenAI(model='gpt-4o')
```

```python
chain = final_prompt | llm | output_parser
```

```python
chain.invoke({'input': '대한민국의 산의 갯수는?'})
```

### CSV 파서

```python
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import CommaSeparatedListOutputParser
```

```python
output_parser = CommaSeparatedListOutputParser()
format_instrunctions = output_parser.get_format_instructions()

print(format_instrunctions)
```

```python
prompt = PromptTemplate(
    template="List five {subject}.\n{format_instrunctions}",
    input_variables=['subject'],
    partial_variables={"format_instrunctions":format_instrunctions}
)
```

```python
llm = ChatOpenAI(model='gpt-4o', temperature=0)
chain = prompt | llm | output_parser

chain.invoke({'subject':'popular Korea cuisine'})
```

### Json 파서
pydantic의 fiel는 BaseModel의 필드를 더 자세하게 정의하고 검증하기 위한 도구입니다. 단순히 타입을 지정하는 것을 넘어서, 데이터의 범위, 형식, 기본값을등을 세밀하게 제어할 수 있게 해줍니다.

```python
from langchain_core.output_parsers import JsonOutputParser
from pydantic import BaseModel, Field

# 자료 구조 정의 pydantic
class CusineRecipe(BaseModel):
    name: str = Field(description = "name of a cuisine")
    recipe: str = Field(description = "recipe to cook the cuisine")

# 출력 파서 정의
output_parser = JsonOutputParser(pydantic_object=CusineRecipe)

format_instrunctions = output_parser.get_format_instructions()

print(format_instrunctions)
```

```python
CR_10 = CusineRecipe(name = '곤', recipe = '없음')
```

```python
CR_10.recipe
```

```python
prompt = PromptTemplate(template= 'Answer the user query. \n {format_instrunctions}\n{query}\n',
                        input_variables=['query'],
                        partial_variables={'format_instrunctions': format_instrunctions})
```

```python
chain = prompt | llm | output_parser
```

```python
chain.invoke({'query' : '씨바꺼 라면 어떻게 요리해'})
```

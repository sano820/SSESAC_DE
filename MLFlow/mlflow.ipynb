# MLFlow

실험의 전체 생명주기(LifeCycle)를 관리하는 오픈소스 플랫폼
- 머신러닝 모델을 만들고 실험하고 배포하기까지의 과정을 더 체계적으로 재현 가능하게 관리해주는 도구
    - 2018년 Databricks에서 개발
    - 다양한 프레임워크에 독립적이고, 설치도 간편한 오픈소스
    - 모델 개발, 배포, 모니터링 등의 전체 프로세스를 하나로 지원함.


#### 머신러닝 생명주기

1. 문제 정의
2. 데이터 수집
3. 데이터 전처리
4. 탐색적 데이터 분석
5. 모델 선택
6. 모델 훈련
7. 모델 평가
8. 모델 배포
9. 모델 모니터링
10. 모델 유지보수 및 과정 반복

## MLFlow 4대 컴포넌트
- Tracking : 실험에서 사용된 파라미터, 메트릭, 아티팩트 등을 기록하고 추적
- Projects : 재사용가능한 형태로 머신러닝 코드를 패키징하여 표준 형식으로 제공
- Models : 다양한 배포 환경에 맞게 모델을 패키징 가능
- Model Registry : 모델의 버전 관리, 상태 추적, 주석 등을 통해 중앙 집중식으로 모델 관리

### MLFlow 주요 구성요소 (1) : Tracking

- Tracking은 머신러닝 실험과 실행을 체계적으로 관리하기 위한 API와 UI를 제공하는 컴포넌트
- Tracking은 실험의 각 실행으로 구성되며, 실험의 각 실행을 ㅗ발생함.
- 이를 통해 실험의 재현성, 투명성, 효율성 향상
- 실험, 파라미터, 성능 지표, 아티팩트를 로깅하고 추적 가능
    - 결과들이 UI 시각화되며 비교 분석도 가능

**로그 및 추적 요소**  
- prameter : 모델 학습에서 사용된 입력 데이터(Hyperparameter)를 키:값 형태로
- Metrics : 모델의 성능 지표(accurafcy, loss 등)를 숫자 형태로.
- Artifact : 모델 파일, 차트, 데이터 파일 등 실행 결과로 생성된 모든 형식의 출력 파일로.
- 코드 버전 : 실행에 사용돈 git commit hash 등을 기록하여 코드 버전 추적

### MLFlow(1) : Tracking

**Tracking URI**  
Tracking Server를 설정
- 로깅함수
    - 파라미터, 메트릭, 아트팩트, 모델 등 머신러닝 실험 중에 발생하는 것들을 추적할 수 있도록 로깅 함수를 제공함.
    - Tracking API의 일부 기능
- Tracking URI
    - Tracking 서버를 설정하는 것으로 사용자가 특정 tracking 서버에 연결 할 수 있게끔 하는 기능
        - 로컬, 원격, 클라우드 플랫폼 등등
- mlflow.set_tracking_uri()
    - 사용할 tracking 서버의 주소를 지정
    - ```mlflow.set_tracking_uri(uri="http://127.0.0.1:5000')```
    - 파라미터
        - uri(str): Tracking Server의 주소

*환경변수로 설정해 놓으면 호출하지 않아도 자동 적용됨  
```export MLFLOW_TRACKING_URI=http://127.0.0.1:5000```

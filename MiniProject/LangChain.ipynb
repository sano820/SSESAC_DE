# LangChain 베이스라인 학습

```python
# 현재 사용자 환경에서 오류가 없다면 이 import 구문을 유지합니다.
from langchain_ollama import ChatOllama 
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

# -------------------- 수정된 부분 --------------------
# 1. Ollama LLM 인스턴스 생성
# 'ollama list'에 표시된 정확한 이름(소문자 및 태그 포함)을 사용합니다.
llm = ChatOllama(model="qwen2.5:latest") 
# ---------------------------------------------------

# 2. 프롬프트 템플릿 정의 (LangChain Expression Language 체인 구성)
prompt = ChatPromptTemplate.from_messages([("system", "당신은 사용자에게 친절하게 대답하는 AI 비서입니다."),("user", "{question}")])

# 3. 출력 파서 정의
output_parser = StrOutputParser()

# 4. 체인 결합 (Prompt -> LLM -> Output Parser)
chain = prompt | llm | output_parser

# 5. 체인 실행 및 결과 출력
question = "한국의 수도는 어디이며, 그 도시에 대해 짧게 설명해 주세요."
print(f"질문: {question}\n")
# invoke() 메서드를 사용하여 체인을 실행합니다.
response = chain.invoke({"question": question})
print("--- Ollama 응답 ---")
print(response)
print("-------------------")
```

# 실습 1: Python 환경 Dockerfile 작성

## 목표

"Hello, Docker!"를 출력하는 간단한 Python 스크립트를 Docker 컨테이너에서 실행합니다.

## Step 1-1: Python 스크립트 작성

먼저 실행할 Python 파일을 작성합니다.

```bash
# 파일 생성
touch hello.py
```

`hello.py` 내용:
```python
print("Hello, Docker!")
print("Python is running inside a container.")
```

---

## Step 1-2: Dockerfile 작성

이제 Dockerfile을 작성합니다. 빈칸을 채워주세요.

```dockerfile
# Dockerfile

# TODO 1: Python 3.9 slim 이미지를 베이스로 사용하세요
# 형식: FROM 이미지명:태그
______________________

# TODO 2: 작업 디렉토리를 /app으로 설정하세요
# 형식: WORKDIR 경로
______________________

# TODO 3: hello.py 파일을 컨테이너의 현재 디렉토리로 복사하세요
# 형식: COPY 원본 대상
______________________

# TODO 4: 컨테이너 시작 시 python hello.py를 실행하세요
# 형식: CMD ["실행파일", "인자"]
______________________
```

<details>
<summary>힌트 1: FROM 명령어</summary>

Docker Hub에서 공식 Python 이미지를 사용합니다.
- `python:3.9` - 전체 이미지
- `python:3.9-slim` - 경량 이미지 (권장)
- `python:3.9-alpine` - 초경량 이미지

</details>

<details>
<summary>힌트 2: WORKDIR 명령어</summary>

컨테이너 내부에서 작업할 기본 디렉토리를 설정합니다.
해당 디렉토리가 없으면 자동으로 생성됩니다.

```dockerfile
WORKDIR /app
```

</details>

<details>
<summary>힌트 3: COPY 명령어</summary>

호스트의 파일을 컨테이너로 복사합니다.
- `.` 은 현재 디렉토리를 의미합니다.
- WORKDIR이 /app이므로 `COPY hello.py .`는 /app/hello.py로 복사됩니다.

</details>

<details>
<summary>모범 답안</summary>

```dockerfile
# Dockerfile

FROM python:3.9-slim

WORKDIR /app

COPY hello.py .

CMD ["python", "hello.py"]
```

</details>

## Step 1-3: 이미지 빌드 및 실행

```bash
# 이미지 빌드
docker build -t hello-python .

# 컨테이너 실행
docker run --rm hello-python
```

# 실습 2: Flask API 서버 Dockerfile 작성

## 목표

간단한 Flask 웹 서버를 Docker 컨테이너에서 실행합니다.

## Step 2-1: 프로젝트 구조 생성

```bash
# 새 폴더 생성
mkdir flask-app
cd flask-app
```

---

## Step 2-2: Flask 애플리케이션 코드 (제공)

아래 코드를 `app.py`로 저장하세요:

```python
# app.py
from flask import Flask, jsonify

app = Flask(__name__)

@app.route('/')
def home():
    return jsonify({
        "message": "Hello from Flask!",
        "status": "running"
    })

@app.route('/health')
def health():
    return jsonify({"status": "healthy"})

if __name__ == '__main__':
    app.run(host='0.0.0.0', port=5000)
```

---

## Step 2-3: requirements.txt 작성

필요한 패키지를 정의합니다.

```
flask==3.0.0
```

---

## Step 2-4: Dockerfile 작성

Flask 앱을 실행하는 Dockerfile을 작성하세요. 빈칸을 채워주세요.

```dockerfile
# Dockerfile

# TODO 1: Python 3.9 slim 이미지 사용
______________________

# TODO 2: 작업 디렉토리 /app 설정
______________________

# TODO 3: requirements.txt를 먼저 복사 (캐싱 최적화)
______________________

# TODO 4: pip로 패키지 설치
# 형식: RUN pip install ...
______________________

# TODO 5: 나머지 소스 코드 복사
# 힌트: 현재 디렉토리의 모든 파일을 복사
______________________

# TODO 6: 컨테이너가 5000번 포트를 사용함을 명시
# 형식: EXPOSE 포트번호
______________________

# TODO 7: Flask 앱 실행
______________________
```

<details>
<summary>힌트 1: requirements.txt를 먼저 복사하는 이유</summary>

Docker는 각 명령어를 레이어로 캐싱합니다.

```dockerfile
# 권장 순서
COPY requirements.txt .    # 의존성 파일만 먼저 복사
RUN pip install ...        # 패키지 설치 (변경 없으면 캐시 사용)
COPY . .                   # 소스 코드 복사
```

소스 코드가 바뀌어도 requirements.txt가 그대로면
pip install 단계는 캐시를 사용하여 빌드가 빨라집니다.

</details>

<details>
<summary>힌트 2: pip install 명령어</summary>

```dockerfile
RUN pip install --no-cache-dir -r requirements.txt
```

- `--no-cache-dir`: pip 캐시를 저장하지 않아 이미지 크기 감소
- `-r requirements.txt`: 파일에서 패키지 목록 읽기

</details>

<details>
<summary>힌트 3: EXPOSE</summary>

`EXPOSE`는 문서화 목적입니다. 실제 포트 매핑은 `docker run -p`로 합니다.

```dockerfile
EXPOSE 5000
```

</details>

<details>
<summary>모범 답안</summary>

```dockerfile
# Dockerfile

FROM python:3.9-slim

WORKDIR /app

COPY requirements.txt .

RUN pip install --no-cache-dir -r requirements.txt

COPY . .

EXPOSE 5000

CMD ["python", "app.py"]
```

</details>

---

## Step 2-5: 이미지 빌드 및 실행

```bash
# 이미지 빌드
docker build -t flask-app .

# 컨테이너 실행 (백그라운드, 포트 매핑)
docker run -d -p 5000:5000 --name my-flask flask-app
```

---

## Step 2-6: 테스트

```bash
# curl로 테스트
curl http://localhost:5000/
curl http://localhost:5000/health

# 또는 브라우저에서 http://localhost:5000 접속
```

**예상 출력**:
```json
{"message":"Hello from Flask!","status":"running"}
{"status":"healthy"}
```

---

## Step 2-7: 정리 및 Git 커밋

```bash
# 실행 중인 컨테이너 중지 및 삭제
docker stop my-flask
docker rm my-flask

# 상위 폴더로 이동
cd ..

# Git 커밋
git add flask-app/
git commit -m "feat: Flask API 서버 Dockerfile 작성"
```

# 실습 3: GitHub Actions 워크플로우 작성

## 목표

GitHub에 Push하면 Docker 이미지가 정상적으로 빌드되고
컨테이너가 실행되는지 자동으로 확인합니다.

## Step 3-1: GitHub 저장소 생성

1. GitHub 접속 → **+** → **New repository**
2. Repository name: `docker-practice`
3. **Public** 선택 → **Create repository**

```bash
# 원격 저장소 연결
git remote add origin https://github.com/YOUR_USERNAME/docker-practice.git
```

---

## Step 3-2: 워크플로우 파일 생성

```bash
# 디렉토리 생성
mkdir -p .github/workflows

# 워크플로우 파일 생성
touch .github/workflows/docker-test.yml
```

---

## Step 3-3: 워크플로우 작성

빈칸을 채워 워크플로우를 완성하세요.

```yaml
# .github/workflows/docker-test.yml

name: Docker Build and Test

# TODO 1: main 브랜치에 push될 때 실행
# 힌트: on: push: branches: [...]
on:
  ______:
    ______: [ main ]

jobs:
  # Job 1: Python Hello World 테스트
  test-hello-python:
    runs-on: ubuntu-latest

    steps:
    # TODO 2: 코드 체크아웃
    # 힌트: uses: actions/checkout@v4
    - name: Checkout code
      ______: ______

    # TODO 3: Docker 이미지 빌드
    # 힌트: run: docker build ...
    - name: Build Docker image
      run: ______ ______ -t hello-python .

    # TODO 4: 컨테이너 실행 및 출력 확인
    - name: Run container
      run: docker run --rm hello-python

  # Job 2: Flask API 테스트
  test-flask-app:
    runs-on: ubuntu-latest

    steps:
    - name: Checkout code
      uses: actions/checkout@v4

    # TODO 5: flask-app 디렉토리에서 이미지 빌드
    # 힌트: working-directory 또는 -f 옵션 사용
    - name: Build Flask Docker image
      run: docker build -t flask-app ./______

    # TODO 6: 백그라운드에서 컨테이너 실행
    # 힌트: -d 옵션 사용
    - name: Run Flask container
      run: docker run ______ -p 5000:5000 --name flask-test flask-app

    # 서버 시작 대기
    - name: Wait for server
      run: sleep 5

    # TODO 7: curl로 헬스체크
    # 힌트: /health 엔드포인트 호출
    - name: Health check
      run: curl --fail http://localhost:5000/______

    # 컨테이너 정리
    - name: Cleanup
      run: docker stop flask-test
```

<details>
<summary>힌트 1: on 트리거 설정</summary>

```yaml
on:
  push:
    branches: [ main ]
```

추가로 pull_request도 트리거할 수 있습니다:
```yaml
on:
  push:
    branches: [ main ]
  pull_request:
    branches: [ main ]
```

</details>

<details>
<summary>힌트 2: actions/checkout</summary>

GitHub Actions에서 소스 코드를 가져오려면 checkout action을 사용합니다.

```yaml
- name: Checkout code
  uses: actions/checkout@v4
```

</details>

<details>
<summary>힌트 3: docker build 명령어</summary>

```bash
docker build -t 이미지명 경로
```

- `-t`: 이미지에 태그(이름) 지정
- `.`: 현재 디렉토리의 Dockerfile 사용
- `./flask-app`: flask-app 폴더의 Dockerfile 사용

</details>

<details>
<summary>힌트 4: curl --fail</summary>

`--fail` 옵션은 HTTP 오류(4xx, 5xx) 시 curl이 실패 종료합니다.
GitHub Actions는 이를 감지하여 워크플로우를 실패로 표시합니다.

</details>

<details>
<summary>모범 답안</summary>

```yaml
# .github/workflows/docker-test.yml

name: Docker Build and Test

on:
  push:
    branches: [ main ]

jobs:
  test-hello-python:
    runs-on: ubuntu-latest

    steps:
    - name: Checkout code
      uses: actions/checkout@v4

    - name: Build Docker image
      run: docker build -t hello-python .

    - name: Run container
      run: docker run --rm hello-python

  test-flask-app:
    runs-on: ubuntu-latest

    steps:
    - name: Checkout code
      uses: actions/checkout@v4

    - name: Build Flask Docker image
      run: docker build -t flask-app ./flask-app

    - name: Run Flask container
      run: docker run -d -p 5000:5000 --name flask-test flask-app

    - name: Wait for server
      run: sleep 5

    - name: Health check
      run: curl --fail http://localhost:5000/health

    - name: Cleanup
      run: docker stop flask-test
```

</details>

# 추가 도전 과제 (선택)

## 도전 1: 멀티 스테이지 빌드

Flask 앱의 Dockerfile을 멀티 스테이지로 최적화해보세요.

<details>
<summary>힌트</summary>

```dockerfile
# 빌드 스테이지
FROM python:3.9 AS builder
WORKDIR /app
COPY requirements.txt .
RUN pip install --user -r requirements.txt

# 실행 스테이지
FROM python:3.9-slim
WORKDIR /app
COPY --from=builder /root/.local /root/.local
COPY . .
ENV PATH=/root/.local/bin:$PATH
CMD ["python", "app.py"]
```

</details>

## 도전 2: docker-compose 추가

flask-app에 docker-compose.yml을 추가해보세요.

<details>
<summary>힌트</summary>

```yaml
# docker-compose.yml
version: '3.8'

services:
  web:
    build: .
    ports:
      - "5000:5000"
    environment:
      - FLASK_ENV=development
```

실행: `docker-compose up --build`

</details>

## 도전 3: GitHub Actions에 캐싱 추가

Docker 빌드 시간을 줄이기 위해 캐싱을 추가해보세요.

<details>
<summary>힌트</summary>

```yaml
- name: Set up Docker Buildx
  uses: docker/setup-buildx-action@v3

- name: Build with cache
  uses: docker/build-push-action@v5
  with:
    context: .
    push: false
    tags: hello-python:latest
    cache-from: type=gha
    cache-to: type=gha,mode=max
```

</details>

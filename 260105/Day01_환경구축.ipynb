## 📌 1. 데이터 엔지니어링이란?

![데이터 엔지니어란 누구일까요](https://drive.google.com/uc?id=1AzwFK74owAO9qudfuLSWQgHW3tYRr22y)

### 🔍 정의

**데이터 엔지니어링**(Data Engineering): 데이터를 수집 → 저장 → 처리 → 제공하는 **시스템을 설계하고 구축**하는 분야

```
📊 원시 데이터 → 🔧 변환/처리 → 💾 저장 → 📈 분석 가능한 형태
```

### 🏗️ 건물로 비유하면

| 역할 | 건축 비유 | 설명 |
|------|----------|------|
| **데이터 엔지니어** | 🏗️ 건축가 + 시공자 | 건물(데이터 시스템)을 설계하고 짓는 사람 |
| **데이터 분석가** | 🏠 인테리어 디자이너 | 완성된 건물에서 공간을 활용하는 사람 |
| **데이터 사이언티스트** | 🔬 연구원 | 건물 안에서 연구/실험하는 사람 |

### ❓ 왜 필요한가?

- 📈 데이터 양의 폭발적 증가 (빅데이터 시대)
- ⚡ 실시간 처리 요구 증가 (추천, 이상 탐지, 알림)
- 🔄 반복적인 작업의 자동화 필요
- 🏢 기업의 데이터 기반 의사결정 증가

```
┌─────────────────────────────────────────────────────────────────────────┐
│                        📥 데이터 소스 (Data Sources)                      │
│         (사용자 이벤트, IoT 센서, 애플리케이션 로그, 외부 API)              │
└─────────────────────────────────────────────────────────────────────────┘
                                    │
                                    ▼
┌─────────────────────────────────────────────────────────────────────────┐
│                         🚀 Apache Kafka (메시지 브로커)                    │
│                    "우체국" - 데이터를 안전하게 전달                        │
└─────────────────────────────────────────────────────────────────────────┘
                    │                               │
         ┌──────────┴──────────┐         ┌─────────┴─────────┐
         ▼                     ▼         ▼                   ▼
┌─────────────────┐   ┌─────────────────┐   ┌─────────────────────────────┐
│ ⚡ Apache Flink  │   │ 🔥 Apache Spark  │   │       🔥 Apache Spark        │
│   (실시간 처리)   │   │  (스트리밍 처리)  │   │       (배치 처리)            │
│ 이상 탐지, CEP   │   │  마이크로 배치    │   │   일일 집계, ETL            │
└─────────────────┘   └─────────────────┘   └─────────────────────────────┘
         │                     │                         │
         └──────────┬──────────┘                         │
                    ▼                                    ▼
┌─────────────────────────────────────────────────────────────────────────┐
│                         💾 데이터 레이크 (S3/MinIO)                        │
│             Raw(원본) → Processed(처리됨) → Curated(정제됨)               │
└─────────────────────────────────────────────────────────────────────────┘
                                    │
                                    ▼
┌─────────────────────────────────────────────────────────────────────────┐
│                    🔍 Elasticsearch + Kibana                             │
│                   검색/분석 + 시각화 대시보드                              │
└─────────────────────────────────────────────────────────────────────────┘
                                    │
                                    ▼
┌─────────────────────────────────────────────────────────────────────────┐
│                           🎯 Apache Airflow                              │
│              워크플로우 자동화 - "매일 새벽 2시에 실행해라"                 │
└─────────────────────────────────────────────────────────────────────────┘
```

### 📦 각 기술 스택의 역할

![데이터의 흐름을 만들고 가공하기 - Kafka, Spark, Flink](https://drive.google.com/uc?id=176TkiDy3lA_cpLZrAYRwPC7nH4E_gy4e)

![어디서든 동일한 환경을 구축하는 기반 기술 - Docker, AWS](https://drive.google.com/uc?id=1FFK8qYDlAyAIoIUC5SayJcwRI8Bptt-c)

| 기술 | 🎯 역할 | 📝 한 줄 설명 | 🔄 비유 |
|------|---------|--------------|---------|
| **Kafka** | 메시지 브로커 | 실시간 데이터를 안전하게 전달 | 우체국 (보내는 쪽과 받는 쪽 연결) |
| **Spark** | 배치/스트리밍 처리 | 대용량 데이터를 빠르게 처리 | 공장 (원자재를 제품으로 가공) |
| **Flink** | 실시간 스트리밍 | 데이터를 즉시 처리 | 컨베이어 벨트 (들어오자마자 처리) |
| **Elasticsearch** | 검색/분석 엔진 | 데이터를 빠르게 검색하고 분석 | 도서관 색인 시스템 |
| **Kibana** | 시각화 도구 | 데이터를 차트/그래프로 표현 | 보고서 작성 도구 |
| **Airflow** | 워크플로우 관리 | 작업 스케줄링 및 자동화 | 비서 (일정 관리) |
| **Docker** | 컨테이너 플랫폼 | 어디서든 동일한 환경으로 실행 | 이사할 때 쓰는 컨테이너 박스 |
| **AWS** | 클라우드 인프라 | 서버, 저장소, 네트워크 제공 | 임대형 데이터센터 |

## 📌 5. 오늘의 목표: 개발 환경 구축

![여정의 시작 - 오늘의 첫걸음](https://drive.google.com/uc?id=1GQEv-48foSxf7quLW4jyE3VLLGHctzbt)

### 🛠️ 오늘 설치/설정할 것들

| 순서 | 항목 | 목적 |
|------|------|------|
| 1 | **WSL2** (Windows Subsystem for Linux) | 윈도우에서 리눅스 사용 |
| 2 | **Docker Desktop** | 컨테이너 환경 구축 |
| 3 | **Git + GitHub** | 버전 관리 및 협업 |
| 4 | **GitHub Actions + Unit Test** | 자동화된 테스트 환경 |

### 💡 왜 이 환경이 필요한가?

```
🖥️ 실무 환경
   └── 대부분 Linux 서버에서 운영
   └── Docker로 배포
   └── Git으로 버전 관리
   └── CI/CD로 자동 테스트/배포

📚 학습 환경 (우리가 구축할 것)
   └── WSL로 Linux 환경 체험
   └── Docker로 실습 환경 통일
   └── GitHub으로 코드 관리
   └── GitHub Actions으로 자동 테스트
```

### 📋 실습 진행 방식

![우리는 이렇게 학습하고 성장합니다 - 빈칸 채우기 방식](https://drive.google.com/uc?id=11MDYirpA0Kb8KOIvFl2zmUcIfhfCBza6)

이 과정에서는 **빈칸 채우기 방식**으로 실습을 진행합니다:

1. 📝 GitHub 저장소 Fork → Clone
2. ✏️ 빈칸(TODO) 부분 코드 작성
3. 🐳 **로컬 Docker 컨테이너**에서 테스트 실행
4. ✅ 테스트 통과 → 다음 Step으로 이동
5. 🔄 Step 완료 시 GitHub Push
6. 🤖 GitHub Actions에서 최종 검증

```
      [Step 1 코드 작성]
             │
             ▼
   ┌─────────────────────┐
   │ 🐳 로컬 Docker 테스트 │  ← docker compose run --rm test
   └─────────────────────┘
             │
       ┌─────┴─────┐
       ▼           ▼
   ✅ 통과       ❌ 실패
       │           │
       ▼           └──→ 코드 수정 후 재테스트
  [다음 Step 진행]
       │
       ▼
   (Step 2, 3, ... 반복)
       │
       ▼
  [모든 Step 완료]
             │
             ▼
      [GitHub Push]
             │
             ▼
   ┌─────────────────────┐
   │ 🤖 GitHub Actions    │  ← 자동 테스트 실행
   └─────────────────────┘
             │
             ▼
   ✅ 최종 통과 확인
```

> 💡 **Tip**: 막혔을 때는 `step-1`, `step-2` 브랜치에서 정답 코드 참고 가능!

<details>
<summary><b>Q1. 데이터 엔지니어와 백엔드 개발자의 차이점은?</b></summary>

**백엔드 개발자**: 애플리케이션의 서버 로직, API 개발에 집중
- 예: 로그인 기능, 결제 시스템, 게시판 CRUD

**데이터 엔지니어**: 데이터 흐름과 처리 파이프라인에 집중
- 예: 로그 수집 시스템, 실시간 추천, 대용량 배치 처리

공통점도 많지만, 데이터 엔지니어는 **데이터 중심 시스템**에 특화됩니다.
</details>

<details>
<summary><b>Q2. Spark와 Flink 둘 다 배워야 하나요?</b></summary>

둘 다 데이터 처리 도구지만 특성이 다릅니다:
- **Spark**: 배치 처리에 강함, 더 널리 사용됨
- **Flink**: 실시간 처리에 강함, 최근 인기 상승 중

현업에서는 둘 다 사용하는 경우가 많아 기본은 알아두는 것이 좋습니다.
</details>

<details>
<summary><b>Q3. 비전공자도 데이터 엔지니어가 될 수 있나요?</b></summary>

네, 충분히 가능합니다. 중요한 것은:
1. Python 기초 역량 (이미 보유!)
2. SQL 활용 능력
3. 시스템적 사고 (데이터 흐름 이해)
4. 포트폴리오 (프로젝트 경험)

이 과정이 바로 이것들을 갖출 수 있도록 설계되어 있습니다.
</details>

<details>
<summary><b>Q4. AI가 다 해주면 왜 배워야 하나요?</b></summary>

AI는 강력한 도구지만, 그 도구를 제어하는 것은 결국 사람입니다:
1. 정확한 요구사항 전달이 필요 → 도메인 지식 필수
* AI에게 무엇을 시킬지 명확히 정의할 수 있어야 원하는 결과가 나옵니다.
2. 결과물 검증 능력이 필요 → 코드 이해력 필수
* AI가 만든 코드가 문법적으로 맞더라도, 논리적 오류나 보안 취약점이 없는지 판단할 수 있어야 합니다.
3. 유지보수와 커스터마이징이 필요 → 소스 코드 장악력 필수
* AI는 초안을 잘 만들어주지만, 실제 서비스에 적용할 때는 내 상황에 맞는 세부 수정이 반드시 따릅니다. 코드를 이해하지 못하면 작은 수정 사항조차 반영할 수 없습니다.
4. 문제 해결 시 디버깅이 필요 → 트러블슈팅 역량 필수
* 에러가 발생했을 때 AI의 답변에만 의존하면 근본적인 원인을 찾지 못해 해결이 더뎌집니다.
5. 면접과 실무에서의 설득 → 원리 이해 필수

## 📌 1. WSL이란?

![왜 Windows 개발자에게 Linux가 필수 스킬이 되었을까요](https://drive.google.com/uc?id=14siJW96mtyXlEuIn9G4y-vS7eR1kR4E6)

![WSL이란 무엇인가](https://drive.google.com/uc?id=1tswIp4t6ecJNQ7Cr4J8rfUAe-AUzR-Yf)

![WSL1 vs WSL2 비교](https://drive.google.com/uc?id=19ZH9cHJ8iJgSBu23zxnWghFzNe032nNO)

### 🔍 정의

**WSL**(Windows Subsystem for Linux): 윈도우에서 리눅스를 실행할 수 있게 해주는 호환성 계층

```
기존 방식:
🖥️ Windows  ──── 🔄 가상머신(무겁다) ──── 🐧 Linux

WSL 방식:
🖥️ Windows  ──── ⚡ WSL(가볍다) ──── 🐧 Linux
```

### 📊 WSL1 vs WSL2

| 특성 | WSL1 | WSL2 |
|------|------|------|
| **커널** | Windows 커널 번역 | 실제 Linux 커널 |
| **파일 시스템 성능** | Windows 파일 빠름 | Linux 파일 빠름 |
| **시스템 호출 호환성** | 부분 지원 | 완전 지원 |
| **Docker 지원** | ❌ 제한적 | ✅ 네이티브 지원 |
| **메모리 사용** | 적음 | 동적 할당 |

> 💡 **우리는 WSL2를 사용합니다** - Docker와의 호환성이 좋고 실제 Linux 환경과 동일

### ❓ 왜 Linux가 필요한가?

| 이유 | 설명 |
|------|------|
| **서버 환경** | 대부분의 서버가 Linux로 운영됨 (AWS, GCP 등) |
| **Docker** | Docker가 Linux 기반 기술 (Windows에서는 WSL2 필요) |
| **오픈소스 도구** | Kafka, Spark 등 대부분 Linux 우선 지원 |
| **개발자 도구** | 패키지 관리, 쉘 스크립트 등이 편리 |

## 📌 3. 트러블슈팅: 자주 발생하는 문제

![예상되는 문제 해결하기 - 흔한 설치 이슈와 해결책](https://drive.google.com/uc?id=1j7o6h6IqVYEg-X-kU-yW_SU93kJV8Loz)

### ❌ 문제 1: "가상화가 BIOS에서 비활성화되어 있습니다"

**해결 방법**:
1. 컴퓨터 재시작 시 BIOS 진입 (F2, F10, DEL 등 - 제조사마다 다름)
2. Advanced → CPU Configuration 찾기
3. **Intel VT-x** 또는 **AMD-V** 활성화
4. 저장 후 재시작

### ❌ 문제 2: "WslRegisterDistribution failed with error: 0x80370102"

**원인**: 가상화 기능이 Windows 기능에서 비활성화됨

**해결**:
```powershell
# Windows 기능 켜기
dism.exe /online /enable-feature /featurename:VirtualMachinePlatform /all /norestart
dism.exe /online /enable-feature /featurename:Microsoft-Windows-Subsystem-Linux /all /norestart
```

### ❌ 문제 3: WSL1으로 설치됨

**해결**: WSL2로 변경
```powershell
# 기본 버전을 WSL2로 설정
wsl --set-default-version 2

# 기존 배포판을 WSL2로 변환
wsl --set-version Ubuntu 2
```

### ✅ 설치 확인

```powershell
wsl -l -v
```

출력 예시:
```
  NAME      STATE           VERSION
* Ubuntu    Running         2
```

![두 세계의 연결 - Windows와 Linux 파일 시스템 접근하기](https://drive.google.com/uc?id=1fxCnDXOkf1ub0nbkJazMIC0FA9XWcL9Z)

```
/                       ← 루트 (최상위)
├── home/               ← 사용자 홈 디렉토리들
│   └── student/        ← 내 홈 디렉토리 (~)
├── etc/                ← 시스템 설정 파일
├── var/                ← 로그, 캐시 등 가변 데이터
│   └── log/            ← 시스템 로그
├── usr/                ← 사용자 프로그램
│   ├── bin/            ← 일반 실행 파일
│   └── local/          ← 사용자 설치 프로그램
├── tmp/                ← 임시 파일
├── opt/                ← 추가 소프트웨어
└── mnt/                ← 마운트 포인트
    └── c/              ← Windows C: 드라이브 (WSL)
```

## ❓ FAQ

<details>
<summary><b>Q1. WSL에서 작업한 파일이 Windows에서 안 보여요</b></summary>

WSL의 파일 시스템과 Windows의 파일 시스템은 분리되어 있습니다.

- WSL 파일 → Windows 탐색기에서 `\\wsl$\Ubuntu` 로 접근
- 또는 `/mnt/c/Users/...` 에서 작업하면 Windows에서 바로 접근 가능

권장: 프로젝트는 `/mnt/c/Users/YourName/projects` 같은 공유 경로에서 작업
</details>

<details>
<summary><b>Q2. 비밀번호를 입력해도 화면에 아무것도 안 나와요</b></summary>

정상입니다! 보안을 위해 비밀번호 입력 시 화면에 표시되지 않습니다.
`*` 도 표시되지 않습니다. 그냥 입력하고 Enter를 누르세요.
</details>

<details>
<summary><b>Q3. sudo가 뭔가요?</b></summary>

`sudo` = Super User Do (관리자 권한으로 실행)

```bash
# 일반 권한 (실패할 수 있음)
apt install something

# 관리자 권한 (비밀번호 필요)
sudo apt install something
```

Windows의 "관리자 권한으로 실행"과 같은 개념입니다.
</details>

<details>
<summary><b>Q4. apt가 뭔가요?</b></summary>

`apt` = Advanced Package Tool (Ubuntu의 패키지 관리자)

```bash
# 패키지 목록 업데이트
sudo apt update

# 패키지 설치
sudo apt install python3

# 패키지 삭제
sudo apt remove python3
```

Windows의 "프로그램 추가/제거"와 비슷하지만 명령어로 실행합니다.
</details>

<details>
<summary><b>Q5. 명령어를 잘못 입력해서 멈췄어요</b></summary>

## 📌 1. Docker란?

![Docker는 무엇이고 왜 혁신적인가요 - VM vs Docker](https://drive.google.com/uc?id=1N-JXLX6EZwrsW32U-u8qypsoowLatfN1)

![Docker 핵심 개념 - 이미지와 컨테이너](https://drive.google.com/uc?id=1rFXhCY5MGOrH1apJQ1nwH_TC9GXuG9Ys)

![Docker 설치 및 첫 실행](https://drive.google.com/uc?id=1y1d924XkEZxCtaKtUaxk_G9TsN_GhyBL)

### 🔍 정의

**Docker**: 애플리케이션을 **컨테이너**라는 격리된 환경에서 실행할 수 있게 해주는 플랫폼

### 🎒 이사할 때 비유

```
기존 방식 (가상머신):
🏠 집 전체를 통째로 옮기기 → 무겁고 느림

Docker 방식 (컨테이너):
📦 필요한 물건만 박스에 담아 옮기기 → 가볍고 빠름
```

### 🆚 가상머신 vs Docker

```
가상머신 (VM)                          Docker (Container)
┌─────────────────────────────┐      ┌─────────────────────────────┐
│  App A   │  App B   │ App C │      │  App A   │  App B   │ App C │
├──────────┼──────────┼───────┤      ├──────────┼──────────┼───────┤
│  Guest   │  Guest   │ Guest │      │   Container Runtime         │
│   OS     │   OS     │  OS   │      │      (Docker Engine)        │
├──────────┴──────────┴───────┤      ├─────────────────────────────┤
│      Hypervisor             │      │         Host OS             │
├─────────────────────────────┤      ├─────────────────────────────┤
│         Host OS             │      │        Hardware             │
├─────────────────────────────┤      └─────────────────────────────┘
│        Hardware             │
└─────────────────────────────┘

OS가 3개 필요 → 무거움                OS 공유 → 가벼움
```

| 특성 | 가상머신 | Docker |
|------|----------|--------|
| **크기** | GB 단위 | MB 단위 |
| **시작 시간** | 분 단위 | 초 단위 |
| **격리 수준** | 완전 격리 | 프로세스 격리 |
| **리소스 사용** | 많음 | 적음 |
| **이식성** | 낮음 | 높음 |

### ❓ 왜 Docker가 필요한가?

#### 😱 "내 컴퓨터에선 되는데..." 문제

```
개발자 A (Windows, Python 3.8)
    "잘 돌아가는데?"
         ↓
서버 (Linux, Python 3.6)
    "에러 발생!"
         ↓
개발자 B (Mac, Python 3.10)
    "나도 에러나는데?"
```

#### 🎉 Docker 해결책

```
개발자 A, B, 서버 모두 동일한 Docker 이미지 사용
    "모든 환경에서 동일하게 동작!"

┌──────────────────────────────────────┐
│         Docker Container             │
│  ┌────────────────────────────────┐  │
│  │  Python 3.9 + 라이브러리       │  │
│  │  + 설정 파일 + 코드            │  │
│  └────────────────────────────────┘  │
└──────────────────────────────────────┘
         ↓           ↓           ↓
      Windows       Mac        Linux
      (동일)       (동일)      (동일)
```

#### 🎯 Docker의 핵심 장점

| 장점 | 설명 |
|------|------|
| **환경 일관성** | 개발/테스트/운영 환경 동일 |
| **빠른 배포** | 이미지만 전달하면 끝 |
| **격리성** | 앱 간 충돌 방지 |
| **확장성** | 컨테이너 복제로 쉽게 확장 |
| **버전 관리** | 이미지 버전으로 롤백 가능 |

## 📌 2. Docker 핵심 개념

### 📦 이미지 (Image) vs 컨테이너 (Container)

```
🎂 이미지 (Image)            📦 컨테이너 (Container)
= 레시피/설계도              = 실제 케이크/건물
= 실행 파일                  = 실행 중인 프로세스
= 읽기 전용                  = 읽기/쓰기 가능
= 변하지 않음                = 상태가 변함

하나의 이미지 → 여러 컨테이너 생성 가능

        ┌─────────────┐
        │  🎂 Image   │
        │  (Python)   │
        └──────┬──────┘
       ┌───────┼───────┐
       ▼       ▼       ▼
    📦 App1  📦 App2  📦 App3
    (실행중) (실행중) (실행중)
```

### 🏪 Docker Hub

**Docker Hub**: Docker 이미지를 저장하고 공유하는 저장소 (앱스토어와 비슷)

```
https://hub.docker.com

공식 이미지 예시:
- python        → Python 실행 환경
- postgres      → PostgreSQL 데이터베이스
- nginx         → 웹 서버
- ubuntu        → Ubuntu Linux
- apache/kafka  → Apache Kafka
```

### 📝 Dockerfile

**Dockerfile**: 이미지를 만드는 레시피 (설정 파일)

```dockerfile
# Dockerfile 예시
FROM python:3.9              # 베이스 이미지
WORKDIR /app                 # 작업 디렉토리
COPY . .                     # 파일 복사
RUN pip install -r req.txt   # 명령 실행
CMD ["python", "app.py"]     # 시작 명령
```

### 📦 컨테이너 관련 명령어

#### 컨테이너 실행 (run)
```bash
# 기본 실행
docker run python:3.9

# 대화형 모드 (-it: interactive + tty)
docker run -it python:3.9

# 백그라운드 실행 (-d: detach)
docker run -d python:3.9

# 이름 지정
docker run --name my-python python:3.9

# 포트 매핑 (-p 호스트:컨테이너)
docker run -p 8080:80 nginx

# 볼륨 마운트 (-v 호스트:컨테이너)
docker run -v /home/user/data:/app/data python:3.9

# 종료 시 자동 삭제 (--rm)
docker run --rm python:3.9
```

#### 주요 옵션 정리

| 옵션 | 의미 | 예시 |
|------|------|------|
| `-it` | 대화형 터미널 | `docker run -it ubuntu bash` |
| `-d` | 백그라운드 실행 | `docker run -d nginx` |
| `--name` | 컨테이너 이름 | `--name my-app` |
| `-p` | 포트 매핑 | `-p 8080:80` |
| `-v` | 볼륨 마운트 | `-v ./data:/data` |
| `--rm` | 종료 시 삭제 | `docker run --rm ...` |
| `-e` | 환경 변수 | `-e MY_VAR=value` |

#### 컨테이너 상태 확인
```bash
# 실행 중인 컨테이너
docker ps

# 모든 컨테이너 (중지된 것 포함)
docker ps -a
```

출력 예시:
```
CONTAINER ID   IMAGE     COMMAND   CREATED         STATUS         PORTS     NAMES
a1b2c3d4e5f6   nginx     ...       5 minutes ago   Up 5 minutes   80/tcp    web-server
```

#### 컨테이너 중지/시작/재시작
```bash
# 중지
docker stop my-container
docker stop a1b2c3d4e5f6   # ID 앞 몇 글자만 입력해도 OK

# 시작 (중지된 컨테이너)
docker start my-container

# 재시작
docker restart my-container
```

#### 컨테이너 삭제
```bash
# 중지된 컨테이너 삭제
docker rm my-container

# 실행 중인 컨테이너 강제 삭제
docker rm -f my-container

# 중지된 모든 컨테이너 삭제
docker container prune
```

#### 컨테이너 내부 접속
```bash
# 실행 중인 컨테이너에 명령 실행
docker exec -it my-container bash

# 특정 명령만 실행
docker exec my-container ls /app
```

#### 로그 확인
```bash
# 로그 보기
docker logs my-container

# 실시간 로그 (follow)
docker logs -f my-container

# 최근 100줄만
docker logs --tail 100 my-container
```

### 🎯 실습 5: PostgreSQL 데이터베이스

```bash
# PostgreSQL 실행
docker run -d \                            # -d: 백그라운드(detach) 모드로 실행
  --name my-postgres \                     # 컨테이너 이름 지정 (나중에 이 이름으로 접근)
  -e POSTGRES_USER=admin \                 # -e: 환경 변수 설정 (DB 사용자명)
  -e POSTGRES_PASSWORD=secret123 \         # DB 비밀번호
  -e POSTGRES_DB=testdb \                  # 생성할 데이터베이스 이름
  -p 5432:5432 \                           # -p: 포트 매핑 (호스트:컨테이너)
  postgres:15                              # 사용할 이미지:태그

# 상태 확인 (실행 중인 컨테이너 목록)
docker ps

# 컨테이너 내부에서 psql 실행
docker exec -it my-postgres psql -U admin -d testdb
# docker exec: 실행 중인 컨테이너에 명령 실행
# -it: 대화형 터미널 모드 (-i: interactive, -t: tty)
# psql -U admin -d testdb: PostgreSQL 클라이언트 실행 (유저: admin, DB: testdb)
```

psql 프롬프트에서:

## 📌 6. Dockerfile 작성하기

### 🔍 Dockerfile이란?

**Dockerfile**: 이미지를 만드는 **설계도/레시피**

```
Dockerfile (레시피)  ──build──→  이미지 (요리)  ──run──→  컨테이너 (먹는 것)
```

### 📝 Dockerfile 기본 문법

```dockerfile
# 주석은 #으로 시작

FROM python:3.9-slim          # 베이스 이미지 (필수, 첫 줄)

WORKDIR /app                  # 작업 디렉토리 설정 (cd와 비슷)

COPY requirements.txt .       # 파일 복사 (호스트 → 컨테이너)

RUN pip install -r requirements.txt  # 명령 실행 (이미지 빌드 시)

COPY . .                      # 나머지 파일 복사

EXPOSE 8000                   # 포트 문서화 (실제 오픈은 -p 옵션)

CMD ["python", "app.py"]      # 컨테이너 시작 시 실행할 명령
```

### 🔑 주요 명령어

| 명령어 | 설명 | 예시 |
|--------|------|------|
| `FROM` | 베이스 이미지 지정 | `FROM python:3.9-slim` |
| `WORKDIR` | 작업 디렉토리 설정 | `WORKDIR /app` |
| `COPY` | 파일 복사 | `COPY . .` |
| `RUN` | 빌드 시 명령 실행 | `RUN pip install pytest` |
| `CMD` | 시작 시 실행 명령 | `CMD ["pytest"]` |
| `ENV` | 환경 변수 설정 | `ENV APP_ENV=production` |
| `EXPOSE` | 포트 문서화 | `EXPOSE 8000` |

> 📖 **공식 문서**: Dockerfile의 모든 명령어와 옵션에 대한 자세한 내용은
> [Dockerfile Reference](https://docs.docker.com/reference/dockerfile/) 참고

### 🎯 실습 6: 첫 번째 Dockerfile 작성

```bash
# 실습 디렉토리 생성
mkdir -p ~/dockerfile-practice
cd ~/dockerfile-practice

# 1. Python 애플리케이션 생성
cat << 'EOF' > app.py
def greet(name):
    return f"Hello, {name}!"

if __name__ == "__main__":
    print(greet("Docker"))
EOF

# 2. 테스트 파일 생성
cat << 'EOF' > test_app.py
from app import greet

def test_greet():
    assert greet("World") == "Hello, World!"
    assert greet("Docker") == "Hello, Docker!"
EOF

# 3. requirements.txt 생성
cat << 'EOF' > requirements.txt
pytest>=7.0.0
EOF

# 4. Dockerfile 생성
cat << 'EOF' > Dockerfile
# Python 3.9 slim 이미지를 베이스로 사용
FROM python:3.9-slim

# 작업 디렉토리 설정
WORKDIR /app

# 의존성 파일 먼저 복사 (캐시 활용)
COPY requirements.txt .

# 의존성 설치
RUN pip install --no-cache-dir -r requirements.txt

# 소스 코드 복사
COPY . .

# 기본 실행 명령: 테스트 실행
CMD ["pytest", "-v"]
EOF
```

### 🔨 이미지 빌드 및 실행

```bash
# 이미지 빌드 (-t: 태그/이름 지정)
docker build -t my-python-app .

# 빌드된 이미지 확인
docker images | grep my-python-app

# 컨테이너 실행 (테스트 실행)
docker run --rm my-python-app

# 예상 출력:
# ========================= test session starts =========================
# collected 1 item
# test_app.py .                                                    [100%]
# ========================== 1 passed in 0.01s ==========================
```

### 💡 빌드 최적화 팁

```dockerfile
# ❌ 비효율적: 코드 변경마다 pip install 다시 실행
COPY . .
RUN pip install -r requirements.txt

# ✅ 효율적: requirements.txt가 변경될 때만 pip install 실행
COPY requirements.txt .
RUN pip install -r requirements.txt
COPY . .
```

### 🔍 Docker Compose란?

**Docker Compose**: 여러 컨테이너를 **한 번에 정의하고 관리**하는 도구

```
docker run 명령어 여러 개  →  docker-compose.yml 하나로 통합
```

### ⚠️ 명령어 버전 안내 (v1 vs v2)

```bash
# ❌ v1 (레거시, deprecated)
docker-compose up

# ✅ v2 (현재 표준, Docker CLI 통합)
docker compose up
```

- **Docker Compose v2**는 Docker CLI에 통합되어 `docker compose` 명령어 사용
- 최신 Docker Desktop에서는 `docker-compose`가 `docker compose`의 **alias**로 설정됨
- 버전 확인: `docker compose version` 또는 `docker-compose --version`

```bash
$ docker-compose --version
Docker Compose version v2.40.3   # ← v2가 alias로 동작
```

> 💡 **권장**: 새 프로젝트에서는 `docker compose` (하이픈 없음) 사용

### 📖 YAML 문법 기초

**YAML**: 설정 파일 형식, **들여쓰기로 구조를 표현**

```yaml
# ✅ 기본 규칙
name: nginx           # 키: 값 (콜론 뒤 공백 필수!)
port: 8080            # 숫자
enabled: true         # 불리언 (true/false)

# ✅ 들여쓰기 = 계층 구조 (스페이스 2칸, 탭 사용 금지!)
services:
  web:                # services의 하위
    image: nginx      # web의 하위

# ✅ 리스트 (하이픈 사용)
ports:
  - "8080:80"
  - "443:443"

# ✅ 딕셔너리
environment:
  USER: admin
  PASSWORD: secret
```

| 규칙 | 예시 | 주의사항 |
|------|------|----------|
| 키: 값 | `name: nginx` | 콜론 뒤 공백 필수 |
| 들여쓰기 | 스페이스 2칸 | 탭 사용 금지 |
| 리스트 | `- item` | 하이픈 + 공백 |
| 주석 | `# 주석` | 샵으로 시작 |

### 📝 docker-compose.yml 기본 구조

```yaml
# docker-compose.yml
services:           # 컨테이너 정의
  web:              # 서비스 이름
    image: nginx    # 사용할 이미지
    ports:
      - "8080:80"   # 포트 매핑

  db:               # 또 다른 서비스
    image: postgres:15
    environment:
      POSTGRES_PASSWORD: secret
```

### 🔑 주요 속성

| 속성 | 설명 | 예시 |
|------|------|------|
| `image` | 사용할 이미지 | `image: python:3.9` |
| `build` | Dockerfile로 빌드 | `build: .` |
| `ports` | 포트 매핑 | `ports: ["8080:80"]` |
| `volumes` | 볼륨 마운트 | `volumes: ["./data:/data"]` |
| `environment` | 환경 변수 | `environment: {KEY: value}` |
| `depends_on` | 의존성 순서 | `depends_on: [db]` |
| `command` | 실행 명령 재정의 | `command: pytest -v` |

> 📖 **공식 문서**: docker-compose.yml의 모든 속성과 설정에 대한 자세한 내용은
> [Compose File Reference](https://docs.docker.com/reference/compose-file/) 참고

### 🎯 실습 7: Docker Compose로 테스트 환경 구성

## 📌 8. 컨테이너에서 테스트 실행하기

### 🎯 왜 컨테이너에서 테스트하는가?

```
❌ 로컬 환경 테스트의 문제
   - "내 컴퓨터에선 되는데..."
   - Python 버전 차이
   - 패키지 버전 충돌
   - 운영체제 차이

✅ 컨테이너 테스트의 장점
   - 모든 개발자가 동일한 환경
   - CI/CD(GitHub Actions)와 동일한 환경
   - 패키지 의존성 격리
   - 재현 가능한 테스트
```

### 🔄 테스트 워크플로우

```
┌─────────────────────────────────────────────────────────────────────┐
│  1. 코드 작성 (빈칸 채우기)                                          │
└─────────────────────────────────────────────────────────────────────┘
                             │
                             ▼
┌─────────────────────────────────────────────────────────────────────┐
│  2. 로컬 컨테이너에서 테스트                                         │
│     $ docker compose run --rm test                                  │
│     → 빠른 피드백, 패키지 설치 불필요                                │
└─────────────────────────────────────────────────────────────────────┘
                             │
                   ┌─────────┴─────────┐
                   ▼                   ▼
             ✅ 통과               ❌ 실패
                   │                   │
                   │                   └── 코드 수정 후 다시 테스트
                   ▼
┌─────────────────────────────────────────────────────────────────────┐
│  3. GitHub에 Push                                                   │
│     $ git add . && git commit -m "..." && git push                  │
└─────────────────────────────────────────────────────────────────────┘
                             │
                             ▼
┌─────────────────────────────────────────────────────────────────────┐
│  4. GitHub Actions 자동 테스트                                       │
│     → 동일한 Docker 환경에서 테스트                                  │
│     → 통과하면 ✅, 실패하면 ❌                                       │
└─────────────────────────────────────────────────────────────────────┘
```

## 📌 10. Docker 리소스 정리 (실무 가이드)

### ⚠️ 왜 정리가 필요한가?

```
여러 프로젝트 테스트 → 이미지/컨테이너 쌓임 → 디스크 용량 부족 😱

특히 문제가 되는 것:
- <none>:<none> 이미지 (댕글링 이미지)
- 중지된 컨테이너
- 사용하지 않는 볼륨
```

### 🔄 안전한 정리 3단계

이미지 삭제 전 **컨테이너를 먼저 정리**해야 함 (사용 중인 이미지는 삭제 불가)

#### Step 1: 중지된 컨테이너 정리

```bash
# 중지된 모든 컨테이너 삭제
docker container prune

# 확인 메시지 출력 → y 입력
```

#### Step 2: 댕글링 이미지 삭제

**댕글링 이미지**: 빌드 과정에서 태그 없이 남은 찌꺼기 (`<none>:<none>`)

```bash
# 이름 없는 이미지들만 삭제
docker image prune
```

#### Step 3: 오래된 이미지 선별 삭제

**필터(Filter)** 기능으로 특정 조건의 이미지만 삭제:

```bash
# 생성된 지 24시간 이상 된 이미지 삭제
docker image prune -a --filter "until=24h"

# 생성된 지 7일(168시간) 이상 된 이미지 삭제
docker image prune -a --filter "until=168h"
```

> ⚠️ `-a` 옵션: 현재 컨테이너가 사용 중이지 않은 **모든** 이미지 삭제
> 나중에 다시 쓸 이미지는 미리 확인 필요

### 💣 한 방에 정리 (가장 강력)

"현재 실행 중인 컨테이너 관련된 것 빼고 다 지워도 된다"면:

```bash
docker system prune -a
```

**삭제 대상**:
- ✅ 중지된 모든 컨테이너
- ✅ 사용되지 않는 네트워크
- ✅ 사용되지 않는 **모든** 이미지
- ✅ 빌드 캐시

> 💡 용량 확보 효과가 가장 큼

### 🔍 삭제 전 용량 확인

```bash
docker system df
```

출력 예시:
```
TYPE            TOTAL     ACTIVE    SIZE      RECLAIMABLE
Images          15        3         5.2GB     4.1GB (78%)
Containers      8         2         120MB     100MB (83%)
Local Volumes   5         2         1.5GB     800MB (53%)
Build Cache     0         0         0B        0B
```

### 🛡️ 애초에 쌓이지 않게 하기 (`--rm`)

테스트나 일회성 작업은 `--rm` 옵션을 붙이면 **종료 시 자동 삭제**:

```bash
# ❌ 컨테이너가 계속 쌓임
docker run python:3.9 python -c "print('hello')"
docker run python:3.9 python -c "print('world')"
# → docker ps -a 하면 중지된 컨테이너 2개 존재

# ✅ 실행 후 자동 삭제 (권장)
docker run --rm python:3.9 python -c "print('hello')"
# → 종료되면 컨테이너 자동 삭제, 깔끔!
```

**Docker Compose도 마찬가지**:

```bash
# ✅ --rm 붙이면 테스트 후 자동 삭제
docker compose run --rm test
```

> 💡 **습관화 추천**: 테스트/디버깅 용도로 컨테이너 실행할 때는 항상 `--rm` 붙이기

### 💡 실무 추천 루틴

```bash
# 1. 용량 확인
docker system df

# 2. 중지된 컨테이너 정리
docker container prune

# 3. 3일 이상 된 이미지 정리
docker image prune -a --filter "until=72h"
```

### ⚡ 특정 이미지만 삭제

```bash
# 특정 이미지 삭제 (이름 또는 ID)
docker rmi python:3.8
docker rmi abc123def456

# 특정 키워드 포함된 이미지 삭제 (예: test)
docker images | grep test | awk '{print $3}' | xargs docker rmi
```

### ⚠️ 주의사항

| 상황 | 해결 방법 |
|------|-----------|
| "image is being used by container" | 해당 컨테이너 먼저 삭제 |
| 중요한 이미지 실수로 삭제 | `docker pull`로 다시 다운로드 |
| 삭제 후에도 용량 부족 | 볼륨도 정리: `docker volume prune` |

## 🔑 핵심 요약

1. **Docker**: 애플리케이션을 컨테이너로 격리하여 실행하는 플랫폼
   - 환경 일관성 확보 ("내 컴퓨터에선 되는데..." 문제 해결)

2. **이미지 vs 컨테이너**:
   - 이미지: 설계도 (읽기 전용)
   - 컨테이너: 실행 중인 인스턴스 (읽기/쓰기)

3. **Dockerfile**: 이미지를 만드는 레시피
   - `FROM`: 베이스 이미지
   - `COPY`: 파일 복사
   - `RUN`: 빌드 시 명령 실행
   - `CMD`: 시작 시 명령

4. **Docker Compose**: 여러 컨테이너를 한 번에 관리
   - `docker compose run --rm test`: 테스트 실행
   - `docker compose up -d`: 전체 서비스 시작
   - v2 명령어 사용 (`docker compose`), `docker-compose`는 alias

5. **테스트 워크플로우**:
   - 코드 작성 → 컨테이너에서 테스트 → GitHub Push → Actions 검증

6. **리소스 정리**: 디스크 용량 관리
   - `docker system df`: 용량 확인
   - `docker container prune`: 중지된 컨테이너 삭제
   - `docker image prune -a --filter "until=72h"`: 오래된 이미지 삭제
   - `docker system prune -a`: 한 방에 전체 정리

<details>
<summary><b>Q2. 포트가 이미 사용 중이라고 나와요</b></summary>

다른 프로그램이 해당 포트를 사용 중입니다.

```bash
# 사용 중인 포트 확인
netstat -ano | findstr :8080

# 다른 포트로 변경
docker run -p 8081:80 nginx
```
</details>

<details>
<summary><b>Q3. 컨테이너를 삭제했는데 데이터가 사라졌어요</b></summary>

컨테이너는 기본적으로 **일시적**(ephemeral)입니다.
데이터를 유지하려면 **볼륨**을 사용해야 합니다.

```bash
# 볼륨 생성
docker volume create mydata

# 볼륨 연결
docker run -v mydata:/app/data ...
```
</details>

<details>
<summary><b>Q4. docker compose run vs docker compose up 차이는?</b></summary>

- `docker compose up`: 모든 서비스를 함께 시작 (웹 서버 + DB 등)
- `docker compose run <서비스>`: 특정 서비스만 실행 (테스트 등 일회성 작업)

```bash
# 테스트 실행 (일회성)
docker compose run --rm test

# 전체 서비스 시작 (백그라운드)
docker compose up -d
```

> 💡 `docker-compose` (하이픈)도 동작하지만, v2 표준인 `docker compose` 사용 권장
</details>

<details>
<summary><b>Q5. 이미지 크기가 너무 커요</b></summary>

**slim** 또는 **alpine** 태그 이미지를 사용하세요.

| 이미지 | 크기 | 설명 |
|--------|------|------|
| `python:3.9` | ~900MB | Debian 기반 전체 버전 (개발 도구 포함) |
| `python:3.9-slim` | ~120MB | Debian 기반 경량 버전 (불필요한 패키지 제거) |
| `python:3.9-alpine` | ~50MB | Alpine Linux 기반 초경량 버전 |

> 💡 **slim vs alpine**
> - **slim**: Debian 기반이라 대부분의 패키지와 호환성 좋음 → 실무에서 권장
> - **alpine**: 가장 작지만 musl libc 사용으로 일부 Python 패키지(numpy, pandas 등) 설치 시 빌드 오류 가능

```bash
docker pull python:3.9-slim
```
</details>

| 구분 | Git | GitHub |
|------|-----|--------|
| **정의** | 버전 관리 도구 | Git 저장소 호스팅 서비스 |

```
[내 컴퓨터]              [인터넷]
   Git        ←────→    GitHub
 (로컬 저장소)          (원격 저장소)
```

## 📌 2. Git 기본 개념

### 🔄 Git의 3가지 영역

```
┌─────────────────┐    ┌─────────────────┐    ┌─────────────────┐
│  Working        │    │   Staging       │    │   Repository    │
│  Directory      │ ─→ │   Area          │ ─→ │   (.git)        │
│  (작업 디렉토리)  │    │   (스테이징)     │    │   (저장소)       │
└─────────────────┘    └─────────────────┘    └─────────────────┘
       │                       │                      │
       │      git add          │      git commit      │
       └───────────────────────┴──────────────────────┘

1. Working Directory: 실제 파일을 수정하는 곳
2. Staging Area: 커밋할 파일을 준비하는 곳
3. Repository: 커밋된 스냅샷이 저장되는 곳
```

### 📸 커밋(Commit)이란?

**커밋**: 특정 시점의 파일 상태를 저장하는 것 (스냅샷)

```
시간 ─────────────────────────────────────────→

    ● commit1      ● commit2      ● commit3
    "초안 작성"     "기능 추가"     "버그 수정"
       │               │               │
       └───────────────┴───────────────┘
            언제든 과거로 돌아갈 수 있음
```

### 🌿 브랜치(Branch)란?

**브랜치**: 독립적인 작업 공간을 만드는 것

```
                    feature/login
                   ┌──● ──● ──●
                  /            \
main ●──●──●──●──●──────────────●──●──●
                  \            /
                   └──● ──●──●
                    feature/signup

- main 브랜치: 안정적인 메인 코드
- feature 브랜치: 새 기능 개발 (완성되면 main에 합침)
```

## 📌 4. Git 기본 명령어

### 📁 저장소 생성/복제

#### 새 저장소 생성
```bash
# 현재 디렉토리를 Git 저장소로 초기화
git init
```

#### 원격 저장소 복제
```bash
# GitHub 저장소를 로컬로 복제
git clone https://github.com/username/repo.git
```

### 📝 파일 추적 및 커밋

#### 상태 확인
```bash
git status
```

출력 예시:
```
On branch main
Changes not staged for commit:
  modified:   app.py          ← 수정됨, 스테이징 안 됨

Untracked files:
  new_file.py                 ← 새 파일, 추적 안 됨
```

#### 스테이징 (add)
```bash
# 특정 파일 추가
git add app.py

# 모든 변경 파일 추가
git add .

# 대화형으로 추가
git add -p
```

#### 커밋 (commit)
```bash
# 메시지와 함께 커밋
git commit -m "feat: 로그인 기능 추가"

# add + commit 한 번에 (추적 중인 파일만)
git commit -am "fix: 버그 수정"
```

### 🔄 원격 저장소 연동

#### 원격 저장소 추가
```bash
git remote add origin https://github.com/username/repo.git
```

#### Push (로컬 → 원격)
```bash
# 처음 push (-u: 업스트림 설정)
git push -u origin main

# 이후 push
git push
```

#### Pull (원격 → 로컬)
```bash
git pull
```

### 📜 이력 확인

#### 커밋 로그
```bash
# 커밋 이력 보기
git log

# 한 줄씩 간략하게
git log --oneline

# 그래프로 보기
git log --oneline --graph --all
```

#### 변경 내용 확인
```bash
# 작업 디렉토리 변경 사항
git diff

# 스테이징된 변경 사항
git diff --staged
```

## 📌 6. GitHub Actions란?

![실습 워크플로우 - 7단계로 완성하는 자동화 파이프라인](https://drive.google.com/uc?id=1TadrnoN19DO1JhEQ0EZzjibznWSrvPjW)

![GitHub Actions 소개](https://drive.google.com/uc?id=1FU5xoM4ApnVYsBjKj_RT2boRQUbYRz3n)

![Unit Test 기초](https://drive.google.com/uc?id=1L2ZDNvCyh4015I6LI-c9TEXJKeToXqcQ)

### 🔍 정의

**GitHub Actions**: GitHub에서 제공하는 **CI/CD**(지속적 통합/배포) 자동화 도구

- **CI**(Continuous Integration): 코드 변경 시 자동으로 테스트/빌드
- **CD**(Continuous Deployment): 테스트 통과 시 자동으로 배포

### 🤖 동작 방식

```
[개발자가 Push]
      │
      ▼
[GitHub Actions 자동 실행]
      │
      ├── ✅ 테스트 실행
      ├── ✅ 코드 품질 검사
      └── ✅ 빌드 확인
      │
      ▼
[결과 표시]
  ✅ 통과 / ❌ 실패
```

### 📋 워크플로우 구조

```yaml
# .github/workflows/test.yml

name: Run Tests              # 워크플로우 이름

on:                          # 트리거 조건
  push:                      # push 할 때 실행
    branches: [ main ]
  pull_request:              # PR 생성 시 실행
    branches: [ main ]

jobs:                        # 작업 정의
  test:                      # 작업 이름
    runs-on: ubuntu-latest   # 실행 환경

    steps:                   # 단계별 실행
    - uses: actions/checkout@v4      # 코드 체크아웃

    - name: Set up Python
      uses: actions/setup-python@v4
      with:
        python-version: '3.9'

    - name: Install dependencies
      run: pip install -r requirements.txt

    - name: Run tests
      run: pytest tests/
```

## 📌 7. Unit Test란?

### 🔍 정의

**Unit Test**(단위 테스트): 코드의 가장 작은 단위(함수, 메서드)가 올바르게 동작하는지 검증하는 테스트

### 🧪 왜 필요한가?

```
❌ 테스트 없이 개발
    "잘 돌아가는 것 같은데?" → 배포 → 💥 버그 발생!

✅ 테스트와 함께 개발
    코드 수정 → 테스트 실행 → ✅ 통과 → 안심하고 배포
```

### 📝 Python pytest 예시

```python
# calculator.py
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b
```

```python
# test_calculator.py
from calculator import add, subtract

def test_add():
    assert add(2, 3) == 5
    assert add(-1, 1) == 0
    assert add(0, 0) == 0

def test_subtract():
    assert subtract(5, 3) == 2
    assert subtract(0, 0) == 0
```
